```python
# ============================================================
# STEP 7: PCA EXPLAINED VARIANCE ANALYSIS
# ============================================================

from sklearn.decomposition import PCA
import numpy as np

print("=" * 70)
print("PCA EXPLAINED VARIANCE ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# Input feature matrix
# ------------------------------------------------------------
X_pca_input = X_processed.copy()

print("\nPCA Input:")
print(f"Samples  : {X_pca_input.shape[0]}")
print(f"Features : {X_pca_input.shape[1]}")

# ------------------------------------------------------------
# Fit PCA with all available components
# ------------------------------------------------------------
pca_full = PCA(
    n_components=X_pca_input.shape[1],
    random_state=42
)

X_pca_full = pca_full.fit_transform(X_pca_input)

explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

# ------------------------------------------------------------
# Display component-wise explained variance
# ------------------------------------------------------------
print("\nCOMPONENT-WISE EXPLAINED VARIANCE")

for i, (individual, cumulative) in enumerate(
    zip(explained_variance, cumulative_variance), 1
):
    print(
        f"PC{i:02d} | "
        f"Explained Variance: {individual:.6f} "
        f"({individual * 100:.2f}%) | "
        f"Cumulative: {cumulative:.6f} "
        f"({cumulative * 100:.2f}%)"
    )

# ------------------------------------------------------------
# Determine components required for 95% variance
# ------------------------------------------------------------
n_components_95 = np.argmax(
    cumulative_variance >= 0.95
) + 1

print("\n95% VARIANCE THRESHOLD")
print(f"Components required : {n_components_95}")
print(
    f"Cumulative explained variance : "
    f"{cumulative_variance[n_components_95 - 1]:.6f} "
    f"({cumulative_variance[n_components_95 - 1] * 100:.2f}%)"
)

# ------------------------------------------------------------
# Variance retained by the selected components
# ------------------------------------------------------------
variance_reduction = (
    1 - (n_components_95 / X_pca_input.shape[1])
) * 100

print("\nDIMENSIONALITY REDUCTION")
print(f"Original dimensions : {X_pca_input.shape[1]}")
print(f"Reduced dimensions  : {n_components_95}")
print(f"Dimensions removed  : {X_pca_input.shape[1] - n_components_95}")
print(f"Dimension reduction : {variance_reduction:.2f}%")

print("\n" + "=" * 70)
print("PCA VARIANCE ANALYSIS COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 8: PCA TRANSFORMATION
# ============================================================

from sklearn.decomposition import PCA
import pandas as pd
import numpy as np

print("=" * 70)
print("PCA DIMENSIONALITY REDUCTION")
print("=" * 70)

# ------------------------------------------------------------
# PCA configuration based on Step 7
# ------------------------------------------------------------
N_COMPONENTS = 16

pca = PCA(
    n_components=N_COMPONENTS,
    random_state=42
)

# ------------------------------------------------------------
# Transform preprocessed feature matrix
# ------------------------------------------------------------
X_pca = pca.fit_transform(X_processed)

# ------------------------------------------------------------
# Create PCA feature names
# ------------------------------------------------------------
pca_columns = [
    f"PC{i:02d}"
    for i in range(1, N_COMPONENTS + 1)
]

X_pca_df = pd.DataFrame(
    X_pca,
    columns=pca_columns,
    index=X_processed.index
)

# ------------------------------------------------------------
# Reattach target
# ------------------------------------------------------------
pca_dataset = pd.concat(
    [X_pca_df, y.rename(target_col)],
    axis=1
)

# ------------------------------------------------------------
# Explained variance
# ------------------------------------------------------------
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("\nPCA INPUT")
print(f"Original feature dimensions : {X_processed.shape[1]}")
print(f"Number of samples           : {X_processed.shape[0]}")

print("\nPCA OUTPUT")
print(f"Reduced feature dimensions  : {X_pca_df.shape[1]}")
print(f"Final PCA dataset shape     : {pca_dataset.shape}")

print("\nEXPLAINED VARIANCE")

for i, (individual, cumulative) in enumerate(
    zip(explained_variance, cumulative_variance), 1
):
    print(
        f"PC{i:02d} | "
        f"Individual: {individual:.6f} "
        f"({individual * 100:.2f}%) | "
        f"Cumulative: {cumulative:.6f} "
        f"({cumulative * 100:.2f}%)"
    )

print("\nPCA SUMMARY")
print(f"Selected components       : {N_COMPONENTS}")
print(
    f"Total variance retained   : "
    f"{cumulative_variance[-1]:.6f} "
    f"({cumulative_variance[-1] * 100:.2f}%)"
)
print(
    f"Variance discarded        : "
    f"{(1 - cumulative_variance[-1]) * 100:.2f}%"
)

# ------------------------------------------------------------
# PCA component matrix dimensions
# ------------------------------------------------------------
print("\nPCA COMPONENT MATRIX")
print(f"Component matrix shape     : {pca.components_.shape}")

# ------------------------------------------------------------
# First five PCA records
# ------------------------------------------------------------
print("\nFIRST 5 PCA REPRESENTATIONS")
print(pca_dataset.head().to_string(index=False))

print("\n" + "=" * 70)
print("PCA TRANSFORMATION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 9: WEIGHTED DISSEMINATION GRAPH CONSTRUCTION
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
from sklearn.metrics.pairwise import cosine_similarity

print("=" * 70)
print("WEIGHTED DISSEMINATION GRAPH CONSTRUCTION")
print("=" * 70)

# ------------------------------------------------------------
# PCA representation used for graph construction
# ------------------------------------------------------------
pca_features = X_pca_df.values

print("\nPCA REPRESENTATION")
print(f"Number of records  : {pca_features.shape[0]}")
print(f"PCA dimensions     : {pca_features.shape[1]}")

# ------------------------------------------------------------
# Node definition
# Each record represents a dissemination pathway state
# ------------------------------------------------------------
node_df = df[
    [
        "event_type",
        "platform",
        "audience",
        "cost",
        "reach",
        "reliability",
        "time_delay",
        "policy_risk"
    ]
].copy()

node_df["node_id"] = np.arange(len(node_df))

# ------------------------------------------------------------
# Edge construction parameters
# ------------------------------------------------------------
SIMILARITY_THRESHOLD = 0.85
MAX_NEIGHBORS = 10

# ------------------------------------------------------------
# PCA cosine similarity
# ------------------------------------------------------------
similarity_matrix = cosine_similarity(pca_features)

np.fill_diagonal(similarity_matrix, 0)

# ------------------------------------------------------------
# Create weighted graph
# ------------------------------------------------------------
G = nx.Graph()

# Add nodes
for idx, row in node_df.iterrows():

    G.add_node(
        int(row["node_id"]),
        event_type=row["event_type"],
        platform=row["platform"],
        audience=row["audience"],
        cost=float(row["cost"]),
        reach=float(row["reach"]),
        reliability=float(row["reliability"]),
        time_delay=float(row["time_delay"]),
        policy_risk=float(row["policy_risk"])
    )

# ------------------------------------------------------------
# Create edges using PCA similarity
# ------------------------------------------------------------
edge_records = []

for i in range(len(node_df)):

    candidate_indices = np.where(
        similarity_matrix[i] >= SIMILARITY_THRESHOLD
    )[0]

    if len(candidate_indices) > MAX_NEIGHBORS:
        candidate_indices = candidate_indices[
            np.argsort(similarity_matrix[i, candidate_indices])[
                -MAX_NEIGHBORS:
            ]
        ]

    for j in candidate_indices:

        if i >= j:
            continue

        similarity = similarity_matrix[i, j]

        # ----------------------------------------------------
        # Dissemination utility:
        # higher reach/reliability is beneficial
        # higher cost/delay/risk is penalized
        # ----------------------------------------------------
        utility = (
            0.30 * node_df.loc[i, "reach"]
            + 0.25 * node_df.loc[i, "reliability"]
            + 0.20 * (1 - node_df.loc[i, "cost"])
            + 0.15 * (1 - node_df.loc[i, "time_delay"])
            + 0.10 * (1 - node_df.loc[i, "policy_risk"])
        )

        edge_weight = similarity * utility

        edge_records.append(
            {
                "source": int(i),
                "target": int(j),
                "similarity": float(similarity),
                "weight": float(edge_weight)
            }
        )

        G.add_edge(
            int(i),
            int(j),
            similarity=float(similarity),
            weight=float(edge_weight)
        )

# ------------------------------------------------------------
# Edge dataframe
# ------------------------------------------------------------
edges_df = pd.DataFrame(edge_records)

# ------------------------------------------------------------
# Graph statistics
# ------------------------------------------------------------
print("\nGRAPH STATISTICS")
print(f"Nodes                 : {G.number_of_nodes()}")
print(f"Edges                 : {G.number_of_edges()}")
print(f"Density               : {nx.density(G):.6f}")

if G.number_of_edges() > 0:

    degrees = np.array(
        [degree for _, degree in G.degree()]
    )

    edge_weights = edges_df["weight"].values

    print(f"Average node degree   : {degrees.mean():.4f}")
    print(f"Maximum node degree   : {degrees.max()}")
    print(f"Minimum node degree   : {degrees.min()}")

    print("\nEDGE WEIGHT STATISTICS")
    print(f"Minimum weight        : {edge_weights.min():.6f}")
    print(f"Maximum weight        : {edge_weights.max():.6f}")
    print(f"Mean weight           : {edge_weights.mean():.6f}")

    print("\nSIMILARITY STATISTICS")
    print(
        f"Minimum similarity    : "
        f"{edges_df['similarity'].min():.6f}"
    )
    print(
        f"Maximum similarity    : "
        f"{edges_df['similarity'].max():.6f}"
    )
    print(
        f"Mean similarity       : "
        f"{edges_df['similarity'].mean():.6f}"
    )

    print("\nFIRST 10 WEIGHTED EDGES")
    print(
        edges_df.head(10).to_string(index=False)
    )

else:

    print("\nNo edges were created.")
    print(
        "The similarity threshold may need adjustment "
        "after inspecting the graph connectivity."
    )

# ------------------------------------------------------------
# Connectivity
# ------------------------------------------------------------
print("\nGRAPH CONNECTIVITY")

if G.number_of_nodes() > 0:

    connected_components = nx.number_connected_components(G)

    print(
        f"Connected components : "
        f"{connected_components}"
    )

    if connected_components == 1:
        print("Graph status          : Connected")
    else:
        largest_component = max(
            nx.connected_components(G),
            key=len
        )

        print(
            f"Largest component    : "
            f"{len(largest_component)} nodes"
        )

print("\n" + "=" * 70)
print("WEIGHTED DISSEMINATION GRAPH CONSTRUCTION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 10: CONNECTIVITY-AWARE GRAPH REFINEMENT
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
from sklearn.metrics.pairwise import cosine_similarity

print("=" * 70)
print("CONNECTIVITY-AWARE GRAPH REFINEMENT")
print("=" * 70)

# ------------------------------------------------------------
# Existing graph
# ------------------------------------------------------------
G_refined = G.copy()

print("\nINITIAL GRAPH")
print(f"Nodes                 : {G_refined.number_of_nodes()}")
print(f"Edges                 : {G_refined.number_of_edges()}")
print(
    f"Connected components  : "
    f"{nx.number_connected_components(G_refined)}"
)

# ------------------------------------------------------------
# PCA similarity matrix
# ------------------------------------------------------------
similarity_matrix = cosine_similarity(
    X_pca_df.values
)

np.fill_diagonal(similarity_matrix, 0)

# ------------------------------------------------------------
# Iteratively connect components
# ------------------------------------------------------------
added_edges = []

while nx.number_connected_components(G_refined) > 1:

    components = list(nx.connected_components(G_refined))

    best_pair = None
    best_similarity = -np.inf

    # Search for the strongest PCA-similarity connection
    # between different connected components.
    for component_i in range(len(components)):

        comp_a = list(components[component_i])

        for component_j in range(
            component_i + 1,
            len(components)
        ):

            comp_b = list(components[component_j])

            for source in comp_a:
                for target in comp_b:

                    similarity = similarity_matrix[
                        source,
                        target
                    ]

                    if similarity > best_similarity:
                        best_similarity = similarity
                        best_pair = (
                            source,
                            target
                        )

    # --------------------------------------------------------
    # Add strongest inter-component edge
    # --------------------------------------------------------
    source, target = best_pair

    source_reach = df.loc[source, "reach"]
    target_reach = df.loc[target, "reach"]

    source_reliability = df.loc[
        source, "reliability"
    ]
    target_reliability = df.loc[
        target, "reliability"
    ]

    source_cost = df.loc[source, "cost"]
    target_cost = df.loc[target, "cost"]

    source_delay = df.loc[
        source, "time_delay"
    ]
    target_delay = df.loc[
        target, "time_delay"
    ]

    source_risk = df.loc[
        source, "policy_risk"
    ]
    target_risk = df.loc[
        target, "policy_risk"
    ]

    avg_reach = (
        source_reach + target_reach
    ) / 2

    avg_reliability = (
        source_reliability
        + target_reliability
    ) / 2

    avg_cost = (
        source_cost + target_cost
    ) / 2

    avg_delay = (
        source_delay + target_delay
    ) / 2

    avg_risk = (
        source_risk + target_risk
    ) / 2

    utility = (
        0.30 * avg_reach
        + 0.25 * avg_reliability
        + 0.20 * (1 - avg_cost)
        + 0.15 * (1 - avg_delay)
        + 0.10 * (1 - avg_risk)
    )

    edge_weight = (
        best_similarity * utility
    )

    G_refined.add_edge(
        int(source),
        int(target),
        similarity=float(best_similarity),
        weight=float(edge_weight),
        refinement_edge=True
    )

    added_edges.append(
        {
            "source": int(source),
            "target": int(target),
            "similarity": float(best_similarity),
            "weight": float(edge_weight)
        }
    )

# ------------------------------------------------------------
# Refined edge dataframe
# ------------------------------------------------------------
refinement_df = pd.DataFrame(
    added_edges
)

# ------------------------------------------------------------
# Final graph statistics
# ------------------------------------------------------------
final_components = nx.number_connected_components(
    G_refined
)

degrees = np.array(
    [degree for _, degree in G_refined.degree()]
)

all_weights = np.array(
    [
        data["weight"]
        for _, _, data
        in G_refined.edges(data=True)
    ]
)

print("\nREFINEMENT RESULTS")
print(
    f"Edges before refinement : "
    f"{G.number_of_edges()}"
)

print(
    f"Edges added             : "
    f"{len(added_edges)}"
)

print(
    f"Edges after refinement  : "
    f"{G_refined.number_of_edges()}"
)

print(
    f"Connected components    : "
    f"{final_components}"
)

print(
    f"Average node degree     : "
    f"{degrees.mean():.4f}"
)

print(
    f"Minimum node degree     : "
    f"{degrees.min()}"
)

print(
    f"Maximum node degree     : "
    f"{degrees.max()}"
)

print(
    f"Graph density           : "
    f"{nx.density(G_refined):.6f}"
)

print("\nFINAL EDGE WEIGHT STATISTICS")
print(
    f"Minimum weight          : "
    f"{all_weights.min():.6f}"
)

print(
    f"Maximum weight          : "
    f"{all_weights.max():.6f}"
)

print(
    f"Mean weight             : "
    f"{all_weights.mean():.6f}"
)

# ------------------------------------------------------------
# Connectivity verification
# ------------------------------------------------------------
print("\nCONNECTIVITY VERIFICATION")

if nx.is_connected(G_refined):
    print("Graph status            : CONNECTED")
    print(
        f"Connected components    : "
        f"{final_components}"
    )
else:
    print("Graph status            : DISCONNECTED")

    component_sizes = sorted(
        [
            len(component)
            for component in nx.connected_components(
                G_refined
            )
        ],
        reverse=True
    )

    print(
        f"Largest component      : "
        f"{component_sizes[0]}"
    )

# ------------------------------------------------------------
# Refinement edge preview
# ------------------------------------------------------------
print("\nFIRST 10 ADDED CONNECTIVITY EDGES")

if len(refinement_df) > 0:
    print(
        refinement_df
        .head(10)
        .to_string(index=False)
    )
else:
    print("No additional edges were required.")

print("\n" + "=" * 70)
print("GRAPH CONNECTIVITY REFINEMENT COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 11: WG-SAGE WEIGHTED NEIGHBORHOOD SAMPLING
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("WG-SAGE WEIGHTED NEIGHBORHOOD SAMPLING")
print("=" * 70)

# ------------------------------------------------------------
# WG-SAGE configuration
# ------------------------------------------------------------
NUM_NODES = G_refined.number_of_nodes()
INPUT_DIM = X_pca_df.shape[1]

SAMPLE_SIZE = 5
RANDOM_STATE = 42

rng = np.random.default_rng(RANDOM_STATE)

# ------------------------------------------------------------
# PCA node features
# ------------------------------------------------------------
node_features = X_pca_df.values.astype(np.float64)

print("\nNODE FEATURE MATRIX")
print(f"Number of nodes       : {NUM_NODES}")
print(f"Input feature size    : {INPUT_DIM}")
print(f"Feature matrix shape  : {node_features.shape}")

# ------------------------------------------------------------
# Weighted neighborhood sampling
# ------------------------------------------------------------
sampled_neighbors = {}
sampled_weights = {}

for node in range(NUM_NODES):

    neighbors = list(
        G_refined.neighbors(node)
    )

    if len(neighbors) == 0:
        sampled_neighbors[node] = []
        sampled_weights[node] = []
        continue

    weights = np.array(
        [
            G_refined[node][neighbor]["weight"]
            for neighbor in neighbors
        ],
        dtype=np.float64
    )

    # --------------------------------------------------------
    # Convert edge weights into sampling probabilities
    # --------------------------------------------------------
    weight_sum = weights.sum()

    if weight_sum > 0:
        probabilities = weights / weight_sum
    else:
        probabilities = np.ones(
            len(neighbors)
        ) / len(neighbors)

    # --------------------------------------------------------
    # Weighted sampling without replacement
    # --------------------------------------------------------
    sample_count = min(
        SAMPLE_SIZE,
        len(neighbors)
    )

    selected = rng.choice(
        len(neighbors),
        size=sample_count,
        replace=False,
        p=probabilities
    )

    selected_neighbors = [
        neighbors[i]
        for i in selected
    ]

    selected_weights = [
        weights[i]
        for i in selected
    ]

    sampled_neighbors[node] = selected_neighbors
    sampled_weights[node] = selected_weights

# ------------------------------------------------------------
# Weighted neighborhood aggregation
# ------------------------------------------------------------
neighbor_aggregates = np.zeros(
    (NUM_NODES, INPUT_DIM),
    dtype=np.float64
)

for node in range(NUM_NODES):

    neighbors = sampled_neighbors[node]
    weights = sampled_weights[node]

    if len(neighbors) == 0:
        continue

    weights = np.array(
        weights,
        dtype=np.float64
    )

    normalized_weights = (
        weights / weights.sum()
    )

    neighbor_features = node_features[
        neighbors
    ]

    neighbor_aggregates[node] = np.sum(
        neighbor_features
        * normalized_weights[:, None],
        axis=0
    )

# ------------------------------------------------------------
# Concatenate self and weighted-neighbor representations
# ------------------------------------------------------------
wgsage_input = np.concatenate(
    [
        node_features,
        neighbor_aggregates
    ],
    axis=1
)

# ------------------------------------------------------------
# Sampling statistics
# ------------------------------------------------------------
neighbor_counts = np.array(
    [
        len(sampled_neighbors[node])
        for node in range(NUM_NODES)
    ]
)

print("\nWEIGHTED SAMPLING STATISTICS")
print(f"Requested neighbors  : {SAMPLE_SIZE}")
print(
    f"Average sampled      : "
    f"{neighbor_counts.mean():.4f}"
)
print(
    f"Minimum sampled      : "
    f"{neighbor_counts.min()}"
)
print(
    f"Maximum sampled      : "
    f"{neighbor_counts.max()}"
)

print("\nAGGREGATION")
print(
    f"Self feature shape   : "
    f"{node_features.shape}"
)
print(
    f"Neighbor aggregate   : "
    f"{neighbor_aggregates.shape}"
)
print(
    f"WG-SAGE input shape  : "
    f"{wgsage_input.shape}"
)

# ------------------------------------------------------------
# Inspect representative nodes
# ------------------------------------------------------------
print("\nREPRESENTATIVE WEIGHTED NEIGHBORHOODS")

for node in [0, 1, 2, 3, 4]:

    neighbors = sampled_neighbors[node]
    weights = sampled_weights[node]

    print(f"\nNode {node}:")
    print(f"Sampled neighbors : {neighbors}")

    if len(weights) > 0:
        print(
            "Edge weights      : "
            + str(
                [
                    round(float(w), 6)
                    for w in weights
                ]
            )
        )

        normalized = (
            np.array(weights)
            / np.sum(weights)
        )

        print(
            "Sampling weights  : "
            + str(
                [
                    round(float(w), 6)
                    for w in normalized
                ]
            )
        )

# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------
print("\nVALIDATION")

print(
    f"NaN values in node features       : "
    f"{np.isnan(node_features).sum()}"
)

print(
    f"NaN values in neighbor aggregates : "
    f"{np.isnan(neighbor_aggregates).sum()}"
)

print(
    f"NaN values in WG-SAGE input       : "
    f"{np.isnan(wgsage_input).sum()}"
)

print("\n" + "=" * 70)
print("WG-SAGE WEIGHTED NEIGHBORHOOD SAMPLING COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 14: DYNACO INITIALIZATION + COMBINED HEURISTIC
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("DYNACO INITIALIZATION + COMBINED HEURISTIC")
print("=" * 70)

# ------------------------------------------------------------
# DynACO parameters
# ------------------------------------------------------------
N_ANTS = 20
N_ITERATIONS = 50

ALPHA = 1.0
BETA = 2.0

INITIAL_PHEROMONE = 1.0

EVAPORATION_MIN = 0.10
EVAPORATION_MAX = 0.50

ELITIST_WEIGHT = 2.0

# Contribution of graph structure and WG-SAGE similarity
GRAPH_WEIGHT = 0.50
EMBEDDING_WEIGHT = 0.50

EPSILON = 1e-8

print("\nDYNACO PARAMETERS")
print(f"Number of ants             : {N_ANTS}")
print(f"Number of iterations       : {N_ITERATIONS}")
print(f"Alpha (pheromone)          : {ALPHA:.2f}")
print(f"Beta (heuristic)           : {BETA:.2f}")
print(f"Initial pheromone          : {INITIAL_PHEROMONE:.2f}")
print(f"Evaporation minimum        : {EVAPORATION_MIN:.2f}")
print(f"Evaporation maximum        : {EVAPORATION_MAX:.2f}")
print(f"Elitist reinforcement      : {ELITIST_WEIGHT:.2f}")

print("\nHEURISTIC CONTRIBUTIONS")
print(f"Graph-weight contribution  : {GRAPH_WEIGHT:.2f}")
print(f"WG-SAGE contribution       : {EMBEDDING_WEIGHT:.2f}")

# ------------------------------------------------------------
# Initialize pheromone matrix
# ------------------------------------------------------------
pheromone_matrix = np.zeros(
    (NUM_NODES, NUM_NODES),
    dtype=np.float64
)

for source, target, data in G_refined.edges(
    data=True
):
    pheromone_matrix[source, target] = (
        INITIAL_PHEROMONE
    )
    pheromone_matrix[target, source] = (
        INITIAL_PHEROMONE
    )

# ------------------------------------------------------------
# Graph edge weights
# ------------------------------------------------------------
graph_weight_matrix = np.zeros(
    (NUM_NODES, NUM_NODES),
    dtype=np.float64
)

for source, target, data in G_refined.edges(
    data=True
):

    weight = float(data["weight"])

    graph_weight_matrix[
        source,
        target
    ] = weight

    graph_weight_matrix[
        target,
        source
    ] = weight

# ------------------------------------------------------------
# Normalize graph edge weights
# ------------------------------------------------------------
graph_max = graph_weight_matrix.max()

if graph_max > 0:
    normalized_graph_weights = (
        graph_weight_matrix / graph_max
    )
else:
    normalized_graph_weights = (
        graph_weight_matrix.copy()
    )

# ------------------------------------------------------------
# Normalize embedding similarity
# ------------------------------------------------------------
embedding_similarity_matrix = (
    embedding_similarity.copy()
)

embedding_max = (
    embedding_similarity_matrix.max()
)

if embedding_max > 0:
    normalized_embedding_similarity = (
        embedding_similarity_matrix
        / embedding_max
    )
else:
    normalized_embedding_similarity = (
        embedding_similarity_matrix.copy()
    )

# ------------------------------------------------------------
# Combined DynACO heuristic
# ------------------------------------------------------------
combined_heuristic = np.zeros(
    (NUM_NODES, NUM_NODES),
    dtype=np.float64
)

edge_mask = graph_weight_matrix > 0

combined_heuristic[edge_mask] = (
    GRAPH_WEIGHT
    * normalized_graph_weights[edge_mask]
    +
    EMBEDDING_WEIGHT
    * normalized_embedding_similarity[edge_mask]
)

# Prevent zero heuristic values on valid edges
combined_heuristic[edge_mask] = np.maximum(
    combined_heuristic[edge_mask],
    EPSILON
)

# ------------------------------------------------------------
# Transition desirability
# ------------------------------------------------------------
heuristic_power = np.zeros_like(
    combined_heuristic
)

heuristic_power[edge_mask] = (
    combined_heuristic[edge_mask] ** BETA
)

transition_desirability = np.zeros_like(
    combined_heuristic
)

transition_desirability[edge_mask] = (
    (
        pheromone_matrix[edge_mask]
        ** ALPHA
    )
    *
    heuristic_power[edge_mask]
)

# ------------------------------------------------------------
# Statistics for valid graph edges
# ------------------------------------------------------------
valid_heuristics = combined_heuristic[
    edge_mask
]

valid_graph_weights = graph_weight_matrix[
    edge_mask
]

valid_embedding_similarities = (
    embedding_similarity_matrix[
        edge_mask
    ]
)

valid_transition = transition_desirability[
    edge_mask
]

print("\nGRAPH EDGE INPUT")

print(
    f"Number of graph edges     : "
    f"{len(valid_graph_weights)}"
)

print(
    f"Graph weight minimum      : "
    f"{valid_graph_weights.min():.6f}"
)

print(
    f"Graph weight maximum      : "
    f"{valid_graph_weights.max():.6f}"
)

print(
    f"Graph weight mean         : "
    f"{valid_graph_weights.mean():.6f}"
)

print("\nWG-SAGE EDGE SIMILARITY")

print(
    f"Similarity minimum        : "
    f"{valid_embedding_similarities.min():.6f}"
)

print(
    f"Similarity maximum        : "
    f"{valid_embedding_similarities.max():.6f}"
)

print(
    f"Similarity mean           : "
    f"{valid_embedding_similarities.mean():.6f}"
)

print("\nCOMBINED HEURISTIC")

print(
    f"Minimum heuristic         : "
    f"{valid_heuristics.min():.6f}"
)

print(
    f"Maximum heuristic         : "
    f"{valid_heuristics.max():.6f}"
)

print(
    f"Mean heuristic            : "
    f"{valid_heuristics.mean():.6f}"
)

print(
    f"Median heuristic          : "
    f"{np.median(valid_heuristics):.6f}"
)

print("\nTRANSITION DESIRABILITY")

print(
    f"Minimum desirability      : "
    f"{valid_transition.min():.6f}"
)

print(
    f"Maximum desirability      : "
    f"{valid_transition.max():.6f}"
)

print(
    f"Mean desirability         : "
    f"{valid_transition.mean():.6f}"
)

# ------------------------------------------------------------
# Representative edge analysis
# ------------------------------------------------------------
print("\nREPRESENTATIVE EDGES")

representative_edges = list(
    G_refined.edges(data=True)
)[:10]

for source, target, data in representative_edges:

    graph_component = (
        GRAPH_WEIGHT
        * normalized_graph_weights[
            source,
            target
        ]
    )

    embedding_component = (
        EMBEDDING_WEIGHT
        * normalized_embedding_similarity[
            source,
            target
        ]
    )

    heuristic_value = combined_heuristic[
        source,
        target
    ]

    print(
        f"Edge {source:03d} -> {target:03d} | "
        f"Graph: {graph_component:.6f} | "
        f"Embedding: {embedding_component:.6f} | "
        f"Heuristic: {heuristic_value:.6f}"
    )

# ------------------------------------------------------------
# Pheromone validation
# ------------------------------------------------------------
valid_pheromone = pheromone_matrix[
    edge_mask
]

print("\nPHEROMONE INITIALIZATION")

print(
    f"Minimum pheromone         : "
    f"{valid_pheromone.min():.6f}"
)

print(
    f"Maximum pheromone         : "
    f"{valid_pheromone.max():.6f}"
)

print(
    f"Mean pheromone            : "
    f"{valid_pheromone.mean():.6f}"
)

print("\nVALIDATION")

print(
    f"NaN heuristic values      : "
    f"{np.isnan(combined_heuristic).sum()}"
)

print(
    f"Infinite heuristic values : "
    f"{np.isinf(combined_heuristic).sum()}"
)

print(
    f"NaN pheromone values      : "
    f"{np.isnan(pheromone_matrix).sum()}"
)

print(
    f"Valid heuristic edges     : "
    f"{np.sum(edge_mask)}"
)

print("\n" + "=" * 70)
print("DYNACO INITIALIZATION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 15: DYNACO TRANSITION PROBABILITY CALCULATION
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("DYNACO TRANSITION PROBABILITY CALCULATION")
print("=" * 70)

# ------------------------------------------------------------
# Transition probability function
# ------------------------------------------------------------
def calculate_transition_probabilities(
    current_node,
    visited_nodes
):
    """
    Calculate DynACO transition probabilities using:

    P(i,j) proportional to
    pheromone(i,j)^ALPHA *
    heuristic(i,j)^BETA

    Only unvisited graph neighbors are considered.
    """

    neighbors = list(
        G_refined.neighbors(current_node)
    )

    candidate_neighbors = [
        neighbor
        for neighbor in neighbors
        if neighbor not in visited_nodes
    ]

    if len(candidate_neighbors) == 0:
        return pd.DataFrame(
            columns=[
                "neighbor",
                "pheromone",
                "heuristic",
                "desirability",
                "probability"
            ]
        )

    records = []

    desirabilities = []

    for neighbor in candidate_neighbors:

        pheromone = pheromone_matrix[
            current_node,
            neighbor
        ]

        heuristic = combined_heuristic[
            current_node,
            neighbor
        ]

        desirability = (
            (pheromone ** ALPHA)
            *
            (heuristic ** BETA)
        )

        desirabilities.append(
            desirability
        )

        records.append(
            {
                "neighbor": int(neighbor),
                "pheromone": float(pheromone),
                "heuristic": float(heuristic),
                "desirability": float(
                    desirability
                )
            }
        )

    desirabilities = np.array(
        desirabilities,
        dtype=np.float64
    )

    total_desirability = (
        desirabilities.sum()
    )

    if total_desirability <= 0:
        probabilities = np.ones(
            len(candidate_neighbors)
        ) / len(candidate_neighbors)
    else:
        probabilities = (
            desirabilities
            / total_desirability
        )

    for i, probability in enumerate(
        probabilities
    ):
        records[i]["probability"] = float(
            probability
        )

    transition_df = pd.DataFrame(
        records
    )

    transition_df = transition_df.sort_values(
        "probability",
        ascending=False
    ).reset_index(drop=True)

    return transition_df


# ------------------------------------------------------------
# Representative starting node
# ------------------------------------------------------------
START_NODE = 0

visited_nodes = {
    START_NODE
}

transition_df = (
    calculate_transition_probabilities(
        START_NODE,
        visited_nodes
    )
)

# ------------------------------------------------------------
# Display transition statistics
# ------------------------------------------------------------
print("\nTRANSITION CONFIGURATION")

print(
    f"Current node            : "
    f"{START_NODE}"
)

print(
    f"Visited nodes           : "
    f"{len(visited_nodes)}"
)

print(
    f"Available neighbors     : "
    f"{G_refined.degree(START_NODE)}"
)

print(
    f"Unvisited candidates    : "
    f"{len(transition_df)}"
)

# ------------------------------------------------------------
# Transition probability statistics
# ------------------------------------------------------------
if len(transition_df) > 0:

    print("\nTRANSITION PROBABILITY STATISTICS")

    print(
        f"Minimum probability    : "
        f"{transition_df['probability'].min():.6f}"
    )

    print(
        f"Maximum probability    : "
        f"{transition_df['probability'].max():.6f}"
    )

    print(
        f"Mean probability       : "
        f"{transition_df['probability'].mean():.6f}"
    )

    print(
        f"Probability sum        : "
        f"{transition_df['probability'].sum():.6f}"
    )

    # --------------------------------------------------------
    # Top transition candidates
    # --------------------------------------------------------
    print("\nTOP TRANSITION CANDIDATES")

    print(
        transition_df
        .head(10)
        .to_string(index=False)
    )

# ------------------------------------------------------------
# Verify transition probability normalization
# ------------------------------------------------------------
print("\nPROBABILITY NORMALIZATION")

if len(transition_df) > 0:

    probability_sum = (
        transition_df["probability"]
        .sum()
    )

    normalization_error = abs(
        probability_sum - 1.0
    )

    print(
        f"Probability sum        : "
        f"{probability_sum:.12f}"
    )

    print(
        f"Normalization error    : "
        f"{normalization_error:.12f}"
    )

    if np.isclose(
        probability_sum,
        1.0,
        atol=1e-10
    ):
        print(
            "Normalization status  : VALID"
        )
    else:
        print(
            "Normalization status  : INVALID"
        )

# ------------------------------------------------------------
# Verify that visited nodes are excluded
# ------------------------------------------------------------
print("\nVISITED-NODE EXCLUSION")

visited_in_candidates = any(
    neighbor in visited_nodes
    for neighbor in transition_df[
        "neighbor"
    ].tolist()
)

print(
    f"Visited node selected  : "
    f"{visited_in_candidates}"
)

# ------------------------------------------------------------
# Global transition validation
# ------------------------------------------------------------
print("\nGLOBAL VALIDATION")

print(
    f"NaN probabilities      : "
    f"{transition_df['probability'].isna().sum()}"
)

print(
    f"Negative probabilities : "
    f"{(
        transition_df['probability'] < 0
    ).sum()}"
)

print(
    f"Maximum probability    : "
    f"{transition_df['probability'].max():.6f}"
)

print(
    f"Minimum probability    : "
    f"{transition_df['probability'].min():.6f}"
)

print("\n" + "=" * 70)
print("DYNACO TRANSITION PROBABILITY CALCULATION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 16: DYNACO ANT PATH CONSTRUCTION
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("DYNACO ANT PATH CONSTRUCTION")
print("=" * 70)

# ------------------------------------------------------------
# Path construction parameters
# ------------------------------------------------------------
PATH_LENGTH = 10
N_ANTS = 20
RANDOM_STATE = 42

rng = np.random.default_rng(RANDOM_STATE)

print("\nPATH CONSTRUCTION PARAMETERS")
print(f"Number of ants           : {N_ANTS}")
print(f"Requested path length    : {PATH_LENGTH}")
print(f"Random seed              : {RANDOM_STATE}")

# ------------------------------------------------------------
# Ant path construction
# ------------------------------------------------------------
def construct_ant_path(
    start_node,
    path_length
):
    """
    Construct one ant pathway using
    DynACO transition probabilities.
    """

    path = [int(start_node)]
    visited = {int(start_node)}

    while len(path) < path_length:

        current_node = path[-1]

        neighbors = list(
            G_refined.neighbors(current_node)
        )

        candidates = [
            neighbor
            for neighbor in neighbors
            if neighbor not in visited
        ]

        # Stop if no feasible unvisited neighbor exists
        if len(candidates) == 0:
            break

        desirabilities = np.array(
            [
                (
                    pheromone_matrix[
                        current_node,
                        neighbor
                    ] ** ALPHA
                )
                *
                (
                    combined_heuristic[
                        current_node,
                        neighbor
                    ] ** BETA
                )
                for neighbor in candidates
            ],
            dtype=np.float64
        )

        total = desirabilities.sum()

        if total <= 0:
            probabilities = np.ones(
                len(candidates)
            ) / len(candidates)
        else:
            probabilities = (
                desirabilities / total
            )

        # Probabilistic transition
        next_node = rng.choice(
            candidates,
            p=probabilities
        )

        path.append(int(next_node))
        visited.add(int(next_node))

    return path


# ------------------------------------------------------------
# Starting-node selection
# ------------------------------------------------------------
start_nodes = rng.choice(
    NUM_NODES,
    size=N_ANTS,
    replace=False
)

# ------------------------------------------------------------
# Construct paths for all ants
# ------------------------------------------------------------
ant_paths = []

for ant_id, start_node in enumerate(
    start_nodes,
    start=1
):

    path = construct_ant_path(
        start_node,
        PATH_LENGTH
    )

    ant_paths.append(path)

# ------------------------------------------------------------
# Path statistics
# ------------------------------------------------------------
path_lengths = np.array(
    [
        len(path)
        for path in ant_paths
    ]
)

unique_path_lengths = np.unique(
    path_lengths
)

print("\nPATH CONSTRUCTION RESULTS")

print(
    f"Paths constructed        : "
    f"{len(ant_paths)}"
)

print(
    f"Minimum path length      : "
    f"{path_lengths.min()}"
)

print(
    f"Maximum path length      : "
    f"{path_lengths.max()}"
)

print(
    f"Average path length      : "
    f"{path_lengths.mean():.4f}"
)

print(
    f"Target path length       : "
    f"{PATH_LENGTH}"
)

# ------------------------------------------------------------
# Full-length path count
# ------------------------------------------------------------
full_length_count = np.sum(
    path_lengths == PATH_LENGTH
)

print(
    f"Full-length paths        : "
    f"{full_length_count}/{N_ANTS}"
)

# ------------------------------------------------------------
# Display all constructed paths
# ------------------------------------------------------------
print("\nCONSTRUCTED ANT PATHS")

for ant_id, path in enumerate(
    ant_paths,
    start=1
):

    print(
        f"Ant {ant_id:02d} | "
        f"Length: {len(path):02d} | "
        f"Path: {path}"
    )

# ------------------------------------------------------------
# Check for repeated nodes
# ------------------------------------------------------------
print("\nPATH UNIQUENESS VALIDATION")

repeated_node_paths = 0

for ant_id, path in enumerate(
    ant_paths,
    start=1
):

    if len(path) != len(set(path)):
        repeated_node_paths += 1

print(
    f"Paths containing repeated nodes : "
    f"{repeated_node_paths}"
)

# ------------------------------------------------------------
# Calculate path transition quality
# ------------------------------------------------------------
path_scores = []

for path in ant_paths:

    if len(path) < 2:
        path_scores.append(0.0)
        continue

    transition_scores = []

    for i in range(len(path) - 1):

        source = path[i]
        target = path[i + 1]

        transition_scores.append(
            combined_heuristic[
                source,
                target
            ]
        )

    path_scores.append(
        np.mean(transition_scores)
    )

path_scores = np.array(
    path_scores
)

print("\nPATH HEURISTIC QUALITY")

print(
    f"Minimum mean heuristic : "
    f"{path_scores.min():.6f}"
)

print(
    f"Maximum mean heuristic : "
    f"{path_scores.max():.6f}"
)

print(
    f"Mean path heuristic    : "
    f"{path_scores.mean():.6f}"
)

# ------------------------------------------------------------
# Best preliminary path
# ------------------------------------------------------------
best_ant_index = np.argmax(
    path_scores
)

best_path = ant_paths[
    best_ant_index
]

print("\nPRELIMINARY BEST PATH")

print(
    f"Best ant               : "
    f"{best_ant_index + 1}"
)

print(
    f"Path length            : "
    f"{len(best_path)}"
)

print(
    f"Mean heuristic score   : "
    f"{path_scores[best_ant_index]:.6f}"
)

print(
    f"Best path              : "
    f"{best_path}"
)

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------
print("\nVALIDATION")

print(
    f"Invalid node IDs       : "
    f"{sum(
        any(
            node < 0 or node >= NUM_NODES
            for node in path
        )
        for path in ant_paths
    )}"
)

print(
    f"Repeated-node paths    : "
    f"{repeated_node_paths}"
)

print(
    f"Valid graph paths      : "
    f"{sum(
        all(
            G_refined.has_edge(
                path[i],
                path[i + 1]
            )
            for i in range(len(path) - 1)
        )
        for path in ant_paths
    )}"
)

print("\n" + "=" * 70)
print("DYNACO ANT PATH CONSTRUCTION COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# SCALABILITY ANALYSIS
# Network Size vs Candidate Edges, Runtime,
# Convergence Iterations and Final Objective
# ============================================================

import time
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# NETWORK SIZES
# ------------------------------------------------------------

network_sizes = [100, 250, 500, 750, 1000]

scalability_results = []

# ------------------------------------------------------------
# RUN SCALABILITY TEST
# ------------------------------------------------------------

for network_size in network_sizes:

    print(f"\nRunning network size: {network_size}")

    # --------------------------------------------------------
    # Select first N nodes from the refined graph
    # --------------------------------------------------------

    selected_nodes = list(range(network_size))

    G_test = G_refined.subgraph(selected_nodes).copy()

    candidate_edges = G_test.number_of_edges()

    # --------------------------------------------------------
    # Start runtime
    # --------------------------------------------------------

    start_time = time.perf_counter()

    # --------------------------------------------------------
    # Initialize pheromone matrix
    # --------------------------------------------------------

    pheromone_test = np.zeros((network_size, network_size), dtype=float)

    for u, v in G_test.edges():
        pheromone_test[u, v] = INITIAL_PHEROMONE
        pheromone_test[v, u] = INITIAL_PHEROMONE

    # --------------------------------------------------------
    # Graph weight matrix
    # --------------------------------------------------------

    graph_weight_test = np.zeros(
        (network_size, network_size),
        dtype=float
    )

    for u, v, data in G_test.edges(data=True):
        weight = float(data.get("weight", 0.0))
        graph_weight_test[u, v] = weight
        graph_weight_test[v, u] = weight

    # --------------------------------------------------------
    # Embedding similarity matrix
    # --------------------------------------------------------

    embedding_test = embedding_similarity[
        :network_size,
        :network_size
    ].copy()

    # --------------------------------------------------------
    # Normalize graph weights
    # --------------------------------------------------------

    edge_values = graph_weight_test[graph_weight_test > 0]

    if len(edge_values) > 0:
        graph_min = edge_values.min()
        graph_max = edge_values.max()

        if graph_max > graph_min:
            normalized_graph_test = (
                (graph_weight_test - graph_min)
                / (graph_max - graph_min)
            )
        else:
            normalized_graph_test = graph_weight_test.copy()
    else:
        normalized_graph_test = graph_weight_test.copy()

    # --------------------------------------------------------
    # Normalize embedding similarity
    # --------------------------------------------------------

    embedding_values = embedding_test[embedding_test > 0]

    if len(embedding_values) > 0:
        emb_min = embedding_values.min()
        emb_max = embedding_values.max()

        if emb_max > emb_min:
            normalized_embedding_test = (
                (embedding_test - emb_min)
                / (emb_max - emb_min)
            )
        else:
            normalized_embedding_test = embedding_test.copy()
    else:
        normalized_embedding_test = embedding_test.copy()

    # --------------------------------------------------------
    # Combined heuristic
    # --------------------------------------------------------

    combined_heuristic_test = (
        GRAPH_WEIGHT * normalized_graph_test
        + EMBEDDING_WEIGHT * normalized_embedding_test
    )

    combined_heuristic_test[graph_weight_test == 0] = 0

    # --------------------------------------------------------
    # DynACO scalability optimization
    # --------------------------------------------------------

    best_score = -np.inf
    best_iteration = 0
    best_path = []

    for iteration in range(1, N_ITERATIONS + 1):

        iteration_scores = []

        # ----------------------------------------------------
        # Construct ant paths
        # ----------------------------------------------------

        for ant in range(N_ANTS):

            start_node = np.random.randint(0, network_size)

            path = [start_node]

            while len(path) < PATH_LENGTH:

                current_node = path[-1]

                neighbors = list(G_test.neighbors(current_node))

                candidates = [
                    n for n in neighbors
                    if n not in path
                ]

                if len(candidates) == 0:
                    break

                desirabilities = []

                for candidate in candidates:

                    pheromone_value = (
                        pheromone_test[current_node, candidate]
                        ** ALPHA
                    )

                    heuristic_value = (
                        combined_heuristic_test[
                            current_node,
                            candidate
                        ]
                        ** BETA
                    )

                    desirability = (
                        pheromone_value
                        * heuristic_value
                    )

                    desirabilities.append(desirability)

                desirabilities = np.asarray(desirabilities)

                total = desirabilities.sum()

                if total <= 0 or not np.isfinite(total):
                    probabilities = np.ones(
                        len(candidates)
                    ) / len(candidates)
                else:
                    probabilities = (
                        desirabilities / total
                    )

                next_node = np.random.choice(
                    candidates,
                    p=probabilities
                )

                path.append(int(next_node))

            # ------------------------------------------------
            # Evaluate pathway objective
            # ------------------------------------------------

            if len(path) >= 2:

                score = evaluate_pathway_objective(path)

                if np.isfinite(score):
                    iteration_scores.append(score)

                    if score > best_score:
                        best_score = score
                        best_path = path.copy()
                        best_iteration = iteration

            elif len(path) == 1:

                score = evaluate_pathway_objective(path)

                if np.isfinite(score):
                    iteration_scores.append(score)

                    if score > best_score:
                        best_score = score
                        best_path = path.copy()
                        best_iteration = iteration

        # ----------------------------------------------------
        # Dynamic evaporation
        # ----------------------------------------------------

        if len(iteration_scores) > 0:

            iteration_best = max(iteration_scores)

            dynamic_evaporation = (
                EVAPORATION_MIN
                + (
                    EVAPORATION_MAX
                    - EVAPORATION_MIN
                )
                * (
                    1
                    - iteration_best
                )
            )

            dynamic_evaporation = np.clip(
                dynamic_evaporation,
                EVAPORATION_MIN,
                EVAPORATION_MAX
            )

            pheromone_test *= (
                1 - dynamic_evaporation
            )

        # ----------------------------------------------------
        # Elitist pheromone reinforcement
        # ----------------------------------------------------

        if len(best_path) >= 2:

            for i in range(len(best_path) - 1):

                u = best_path[i]
                v = best_path[i + 1]

                if G_test.has_edge(u, v):

                    pheromone_test[u, v] += (
                        ELITIST_WEIGHT * best_score
                    )

                    pheromone_test[v, u] = (
                        pheromone_test[u, v]
                    )

    # --------------------------------------------------------
    # Runtime
    # --------------------------------------------------------

    runtime = time.perf_counter() - start_time

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    scalability_results.append({
        "Network size (nodes)": network_size,
        "Candidate edges": candidate_edges,
        "Runtime (s)": runtime,
        "Convergence iterations": best_iteration,
        "Final objective value": best_score
    })

# ============================================================
# RESULTS TABLE
# ============================================================

scalability_df = pd.DataFrame(
    scalability_results
)

print("\n")
print("=" * 75)
print("SCALABILITY RESULTS")
print("=" * 75)

print(
    scalability_df.to_string(
        index=False,
        formatters={
            "Runtime (s)": "{:.4f}".format,
            "Final objective value": "{:.6f}".format
        }
    )
)
```

```python
# ============================================================
# MODEL COMPARISON
# SVM | Random Forest | XGBoost | MLP | ACO | ACGNN | DynACO | WG-SAGE
# OUTPUT: PRECISION, RECALL, F1-SCORE ONLY
# ============================================================

import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import precision_score, recall_score, f1_score

# XGBoost
from xgboost import XGBClassifier

# PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F

# ------------------------------------------------------------
# REPRODUCIBILITY
# ------------------------------------------------------------

SEED = 42

np.random.seed(SEED)
torch.manual_seed(SEED)

# ------------------------------------------------------------
# DATA
# ------------------------------------------------------------
# Your notebook contains:
# X_processed -> 21 features
# df["selected_path"] -> binary target

X_base = X_processed.copy().astype(float)
y_base = df["selected_path"].astype(int).to_numpy()

X = X_base.to_numpy()

# ------------------------------------------------------------
# TRAIN / TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_base,
    test_size=0.20,
    random_state=SEED,
    stratify=y_base
)

# ------------------------------------------------------------
# METRIC FUNCTION
# ------------------------------------------------------------

def evaluate_model(name, y_true, y_pred):

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    print(
        f"{name:<18} "
        f"Precision: {precision:.6f}   "
        f"Recall: {recall:.6f}   "
        f"F1-Score: {f1:.6f}"
    )

    return {
        "Model": name,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    }


results = []


# ============================================================
# 1. SVM
# ============================================================

svm_model = Pipeline([
    ("scaler", StandardScaler()),
    (
        "classifier",
        SVC(
            kernel="rbf",
            C=10,
            gamma="scale",
            random_state=SEED
        )
    )
])

svm_model.fit(X_train, y_train)

y_pred_svm = svm_model.predict(X_test)

results.append(
    evaluate_model(
        "SVM",
        y_test,
        y_pred_svm
    )
)


# ============================================================
# 2. RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    random_state=SEED,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        y_pred_rf
    )
)


# ============================================================
# 3. XGBOOST
# ============================================================

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.90,
    colsample_bytree=0.90,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=SEED,
    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train
)

y_pred_xgb = xgb_model.predict(X_test)

results.append(
    evaluate_model(
        "XGBoost",
        y_test,
        y_pred_xgb
    )
)


# ============================================================
# 4. MLP
# ============================================================

mlp_model = Pipeline([
    ("scaler", StandardScaler()),
    (
        "classifier",
        MLPClassifier(
            hidden_layer_sizes=(128, 64, 32),
            activation="relu",
            solver="adam",
            alpha=0.0001,
            learning_rate_init=0.001,
            max_iter=500,
            random_state=SEED,
            early_stopping=True,
            validation_fraction=0.15
        )
    )
])

mlp_model.fit(
    X_train,
    y_train
)

y_pred_mlp = mlp_model.predict(X_test)

results.append(
    evaluate_model(
        "MLP",
        y_test,
        y_pred_mlp
    )
)


# ============================================================
# 5. ACO
# ============================================================
# ACO is used as a feature-selection optimizer.
# The selected features are then classified using a lightweight
# nearest-centroid decision rule.

def aco_feature_selection(
    X_train,
    y_train,
    n_ants=20,
    n_iterations=30,
    evaporation=0.20,
    alpha=1.0,
    beta=2.0
):

    n_features = X_train.shape[1]

    pheromone = np.ones(n_features)

    # Feature relevance based on absolute correlation
    relevance = np.zeros(n_features)

    for j in range(n_features):
        feature = X_train[:, j]

        if np.std(feature) > 0:
            relevance[j] = abs(
                np.corrcoef(feature, y_train)[0, 1]
            )

    relevance = np.nan_to_num(relevance)

    if relevance.max() > 0:
        relevance = (
            relevance / relevance.max()
        )

    best_subset = None
    best_score = -np.inf

    rng = np.random.default_rng(SEED)

    for _ in range(n_iterations):

        for _ in range(n_ants):

            probability = (
                (pheromone ** alpha) *
                ((relevance + 1e-6) ** beta)
            )

            probability = (
                probability /
                probability.sum()
            )

            # Select approximately 50% features
            n_select = max(
                2,
                int(n_features * 0.50)
            )

            subset = rng.choice(
                n_features,
                size=n_select,
                replace=False,
                p=probability
            )

            subset = np.sort(subset)

            # Evaluate using centroid separation
            class0 = X_train[
                y_train == 0
            ][:, subset]

            class1 = X_train[
                y_train == 1
            ][:, subset]

            if len(class0) == 0 or len(class1) == 0:
                continue

            c0 = class0.mean(axis=0)
            c1 = class1.mean(axis=0)

            separation = np.linalg.norm(
                c0 - c1
            )

            if separation > best_score:
                best_score = separation
                best_subset = subset.copy()

            pheromone *= (
                1 - evaporation
            )

            pheromone[subset] += separation + 1e-6

    if best_subset is None:
        best_subset = np.arange(n_features)

    return best_subset


aco_features = aco_feature_selection(
    X_train,
    y_train
)

X_train_aco = X_train[:, aco_features]
X_test_aco = X_test[:, aco_features]

# Centroid classifier after ACO feature selection

centroid_0 = X_train_aco[
    y_train == 0
].mean(axis=0)

centroid_1 = X_train_aco[
    y_train == 1
].mean(axis=0)

dist0 = np.linalg.norm(
    X_test_aco - centroid_0,
    axis=1
)

dist1 = np.linalg.norm(
    X_test_aco - centroid_1,
    axis=1
)

y_pred_aco = (
    dist1 < dist0
).astype(int)

results.append(
    evaluate_model(
        "ACO",
        y_test,
        y_pred_aco
    )
)


# ============================================================
# GRAPH PREPARATION
# ============================================================

# Use the existing refined graph if available.
# Otherwise construct a k-nearest-neighbor graph.

try:

    graph = G_refined

    n_nodes = len(X)

    A = np.zeros(
        (n_nodes, n_nodes),
        dtype=np.float32
    )

    for u, v, data in graph.edges(
        data=True
    ):

        weight = float(
            data.get("weight", 1.0)
        )

        A[int(u), int(v)] = weight
        A[int(v), int(u)] = weight

except Exception:

    from sklearn.neighbors import kneighbors_graph

    A = kneighbors_graph(
        X,
        n_neighbors=6,
        mode="distance",
        include_self=False
    ).toarray()

    A = np.exp(-A)

    A[A == 1] = 0

# ------------------------------------------------------------
# NORMALIZE ADJACENCY
# ------------------------------------------------------------

A = A + np.eye(
    A.shape[0],
    dtype=np.float32
)

degree = A.sum(axis=1)

degree_inv_sqrt = 1.0 / np.sqrt(
    degree + 1e-8
)

A_norm = (
    degree_inv_sqrt[:, None]
    * A
    * degree_inv_sqrt[None, :]
)

A_tensor = torch.tensor(
    A_norm,
    dtype=torch.float32
)

X_tensor = torch.tensor(
    X,
    dtype=torch.float32
)

y_tensor = torch.tensor(
    y_base,
    dtype=torch.long
)

# ------------------------------------------------------------
# NODE TRAIN / TEST MASK
# ------------------------------------------------------------

indices = np.arange(
    len(y_base)
)

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.20,
    random_state=SEED,
    stratify=y_base
)

train_idx = torch.tensor(
    train_idx,
    dtype=torch.long
)

test_idx = torch.tensor(
    test_idx,
    dtype=torch.long
)


# ============================================================
# 6. ACGNN
# ============================================================

class ACGNNLayer(nn.Module):

    def __init__(
        self,
        in_features,
        out_features
    ):

        super().__init__()

        self.weight = nn.Parameter(
            torch.randn(
                in_features,
                out_features
            ) * 0.05
        )

        self.bias = nn.Parameter(
            torch.zeros(out_features)
        )

        self.attention = nn.Linear(
            out_features,
            1
        )

    def forward(
        self,
        X,
        A
    ):

        H = torch.matmul(
            X,
            self.weight
        ) + self.bias

        H = torch.matmul(
            A,
            H
        )

        attention_score = torch.sigmoid(
            self.attention(H)
        )

        H = H * attention_score

        return F.relu(H)


class ACGNN(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes
    ):

        super().__init__()

        self.layer1 = ACGNNLayer(
            input_dim,
            hidden_dim
        )

        self.layer2 = ACGNNLayer(
            hidden_dim,
            hidden_dim
        )

        self.output = nn.Linear(
            hidden_dim,
            num_classes
        )

    def forward(
        self,
        X,
        A
    ):

        H = self.layer1(X, A)

        H = F.dropout(
            H,
            p=0.20,
            training=self.training
        )

        H = self.layer2(H, A)

        return self.output(H)


acgnn = ACGNN(
    input_dim=X.shape[1],
    hidden_dim=64,
    num_classes=2
)

optimizer = torch.optim.Adam(
    acgnn.parameters(),
    lr=0.005,
    weight_decay=1e-4
)

loss_function = nn.CrossEntropyLoss()

for epoch in range(150):

    acgnn.train()

    optimizer.zero_grad()

    output = acgnn(
        X_tensor,
        A_tensor
    )

    loss = loss_function(
        output[train_idx],
        y_tensor[train_idx]
    )

    loss.backward()

    optimizer.step()


acgnn.eval()

with torch.no_grad():

    output = acgnn(
        X_tensor,
        A_tensor
    )

    y_pred_acgnn = (
        output[test_idx]
        .argmax(dim=1)
        .cpu()
        .numpy()
    )

y_test_graph = (
    y_tensor[test_idx]
    .cpu()
    .numpy()
)

results.append(
    evaluate_model(
        "ACGNN",
        y_test_graph,
        y_pred_acgnn
    )
)


# ============================================================
# 7. DYNACO
# ============================================================
# DynACO uses both graph connectivity and feature similarity
# to obtain an optimized node representation.

def dynaco_node_selection(
    X,
    A,
    y,
    n_ants=20,
    n_iterations=30
):

    n_nodes = len(X)

    # Feature similarity
    norm_X = np.linalg.norm(
        X,
        axis=1,
        keepdims=True
    )

    X_norm = X / (
        norm_X + 1e-8
    )

    similarity = (
        X_norm @ X_norm.T
    )

    similarity = np.clip(
        similarity,
        0,
        1
    )

    # Combined heuristic
    graph_component = A.copy()

    if graph_component.max() > 0:

        graph_component = (
            graph_component /
            graph_component.max()
        )

    heuristic = (
        0.50 * graph_component +
        0.50 * similarity
    )

    pheromone = (
        A + 1e-6
    )

    best_path = None
    best_score = -np.inf

    rng = np.random.default_rng(
        SEED
    )

    for _ in range(n_iterations):

        for _ in range(n_ants):

            start = rng.integers(
                0,
                n_nodes
            )

            path = [int(start)]

            target_length = min(
                50,
                n_nodes
            )

            while len(path) < target_length:

                current = path[-1]

                candidates = np.where(
                    A[current] > 0
                )[0]

                candidates = [
                    int(c)
                    for c in candidates
                    if int(c) not in path
                ]

                if not candidates:
                    break

                desirability = np.array([
                    (
                        pheromone[
                            current,
                            c
                        ] ** 1.0
                    ) *
                    (
                        max(
                            heuristic[
                                current,
                                c
                            ],
                            1e-8
                        ) ** 2.0
                    )
                    for c in candidates
                ])

                if desirability.sum() <= 0:

                    probability = np.ones(
                        len(candidates)
                    ) / len(candidates)

                else:

                    probability = (
                        desirability /
                        desirability.sum()
                    )

                next_node = rng.choice(
                    candidates,
                    p=probability
                )

                path.append(
                    int(next_node)
                )

            if len(path) > 1:

                score = 0.0

                for i in range(
                    len(path) - 1
                ):

                    u = path[i]
                    v = path[i + 1]

                    score += heuristic[
                        u,
                        v
                    ]

                score /= (
                    len(path) - 1
                )

                if score > best_score:

                    best_score = score
                    best_path = path.copy()

                # pheromone update

                for i in range(
                    len(path) - 1
                ):

                    u = path[i]
                    v = path[i + 1]

                    pheromone[
                        u,
                        v
                    ] += score

                    pheromone[
                        v,
                        u
                    ] += score

        pheromone *= 0.80

    if best_path is None:

        best_path = list(
            np.argsort(
                np.sum(
                    heuristic,
                    axis=1
                )
            )[-50:]
        )

    return np.array(
        best_path,
        dtype=int
    )


dynaco_nodes = dynaco_node_selection(
    X,
    A_norm,
    y_base
)

# Train a classifier using optimized DynACO nodes

dynaco_train_nodes = np.intersect1d(
    dynaco_nodes,
    train_idx.numpy()
)

dynaco_test_nodes = np.intersect1d(
    dynaco_nodes,
    test_idx.numpy()
)

# If too few test nodes are selected,
# use the full test set with DynACO-derived features.

if len(dynaco_train_nodes) < 5:

    dynaco_train_nodes = train_idx.numpy()

if len(dynaco_test_nodes) < 5:

    dynaco_test_nodes = test_idx.numpy()

# DynACO-enhanced representation
dynaco_features = (
    0.50 * X +
    0.50 * (
        A_norm @ X
    )
)

X_dyn_train = (
    dynaco_features[
        dynaco_train_nodes
    ]
)

y_dyn_train = (
    y_base[
        dynaco_train_nodes
    ]
)

X_dyn_test = (
    dynaco_features[
        dynaco_test_nodes
    ]
)

y_dyn_test = (
    y_base[
        dynaco_test_nodes
    ]
)

dynaco_classifier = Pipeline([
    ("scaler", StandardScaler()),
    (
        "classifier",
        SVC(
            kernel="rbf",
            C=10,
            gamma="scale",
            random_state=SEED
        )
    )
])

dynaco_classifier.fit(
    X_dyn_train,
    y_dyn_train
)

y_pred_dynaco = (
    dynaco_classifier.predict(
        X_dyn_test
    )
)

results.append(
    evaluate_model(
        "DynACO",
        y_dyn_test,
        y_pred_dynaco
    )
)


# ============================================================
# 8. WG-SAGE
# ============================================================
# Weighted GraphSAGE-style aggregation using the existing
# weighted graph information.

class WGSAGELayer(nn.Module):

    def __init__(
        self,
        in_features,
        out_features
    ):

        super().__init__()

        self.self_linear = nn.Linear(
            in_features,
            out_features
        )

        self.neighbor_linear = nn.Linear(
            in_features,
            out_features
        )

    def forward(
        self,
        X,
        A
    ):

        # Self representation
        self_part = self.self_linear(
            X
        )

        # Weighted neighborhood aggregation
        neighbor_part = (
            self.neighbor_linear(
                torch.matmul(A, X)
            )
        )

        return F.relu(
            self_part +
            neighbor_part
        )


class WGSAGE(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim,
        num_classes
    ):

        super().__init__()

        self.layer1 = WGSAGELayer(
            input_dim,
            hidden_dim
        )

        self.layer2 = WGSAGELayer(
            hidden_dim,
            hidden_dim
        )

        self.output = nn.Linear(
            hidden_dim,
            num_classes
        )

    def forward(
        self,
        X,
        A
    ):

        H = self.layer1(
            X,
            A
        )

        H = F.dropout(
            H,
            p=0.20,
            training=self.training
        )

        H = self.layer2(
            H,
            A
        )

        return self.output(H)


wgsage = WGSAGE(
    input_dim=X.shape[1],
    hidden_dim=64,
    num_classes=2
)

optimizer = torch.optim.Adam(
    wgsage.parameters(),
    lr=0.005,
    weight_decay=1e-4
)

loss_function = nn.CrossEntropyLoss()

for epoch in range(150):

    wgsage.train()

    optimizer.zero_grad()

    output = wgsage(
        X_tensor,
        A_tensor
    )

    loss = loss_function(
        output[train_idx],
        y_tensor[train_idx]
    )

    loss.backward()

    optimizer.step()


wgsage.eval()

with torch.no_grad():

    output = wgsage(
        X_tensor,
        A_tensor
    )

    y_pred_wgsage = (
        output[test_idx]
        .argmax(dim=1)
        .cpu()
        .numpy()
    )

results.append(
    evaluate_model(
        "WG-SAGE",
        y_test_graph,
        y_pred_wgsage
    )
)


# ============================================================
# FINAL RESULTS
# ============================================================

results_df = pd.DataFrame(
    results
)

print("\n")
print(results_df.to_string(index=False))
```

```python
import os
import time
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.preprocessing import MinMaxScaler

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

OUTPUT_DIR = "DynACO_WG_SAGE_Sensitivity_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

SEED = 42

N_ANTS = 20
N_ITERATIONS = 50
PATH_LENGTH = 10

INITIAL_PHEROMONE = 1.0
EVAPORATION_MIN = 0.10
EVAPORATION_MAX = 0.50

ALPHA = 1.0
BETA = 2.0
ELITE_REINFORCEMENT = 2.0

# ------------------------------------------------------------
# 2. VALIDATE INPUT DATA
# ------------------------------------------------------------

required_columns = [
    "reach",
    "reliability",
    "cost",
    "time_delay",
    "policy_risk"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing from df: {missing_columns}"
    )

if "G_refined" not in globals():
    raise NameError(
        "G_refined is unavailable. Run the graph-construction "
        "and graph-refinement cells first."
    )

if len(df) == 0:
    raise ValueError("The input dataset is empty.")

if G_refined.number_of_nodes() == 0:
    raise ValueError("G_refined contains no nodes.")

if G_refined.number_of_edges() == 0:
    raise ValueError(
        "G_refined contains no edges. "
        "Pathway optimization requires graph connectivity."
    )

# This experiment assumes graph node IDs correspond to
# zero-based row positions in df, as in the existing notebook.

valid_nodes = set(range(len(df)))

if not set(G_refined.nodes()).issubset(valid_nodes):
    raise ValueError(
        "Graph node IDs do not match df row positions. "
        "Check the node-ID mapping before running this experiment."
    )

if not nx.is_connected(G_refined):
    print(
        "WARNING: G_refined is disconnected. "
        "Paths may terminate before reaching the requested length."
    )

print(f"Dataset records : {len(df)}")
print(f"Graph nodes     : {G_refined.number_of_nodes()}")
print(f"Graph edges     : {G_refined.number_of_edges()}")

# ------------------------------------------------------------
# 3. NORMALIZE OBJECTIVE COMPONENTS
# ------------------------------------------------------------

objective_data = df[required_columns].copy()

for col in required_columns:
    objective_data[col] = pd.to_numeric(
        objective_data[col],
        errors="coerce"
    )

if objective_data.isna().any().any():
    raise ValueError(
        "Objective columns contain missing or non-numeric values. "
        "Resolve these values before running sensitivity analysis."
    )

if not np.isfinite(objective_data.to_numpy()).all():
    raise ValueError(
        "Objective columns contain infinite or non-finite values."
    )

scaler = MinMaxScaler()

normalized_values = scaler.fit_transform(objective_data)

normalized_df = pd.DataFrame(
    normalized_values,
    columns=required_columns,
    index=np.arange(len(df))
)

# Convert cost-like attributes to benefit-oriented scores.
normalized_df["cost_benefit"] = (
    1.0 - normalized_df["cost"]
)

normalized_df["delay_benefit"] = (
    1.0 - normalized_df["time_delay"]
)

normalized_df["risk_benefit"] = (
    1.0 - normalized_df["policy_risk"]
)

# ------------------------------------------------------------
# 4. DEFINE SIX WEIGHT CONFIGURATIONS
# ------------------------------------------------------------

weight_configurations = {
    "Equal weights": {
        "reach": 0.20,
        "reliability": 0.20,
        "cost": 0.20,
        "delay": 0.20,
        "policy_risk": 0.20
    },

    "Cost-priority": {
        "reach": 0.10,
        "reliability": 0.10,
        "cost": 0.50,
        "delay": 0.15,
        "policy_risk": 0.15
    },

    "Reach-priority": {
        "reach": 0.55,
        "reliability": 0.15,
        "cost": 0.10,
        "delay": 0.10,
        "policy_risk": 0.10
    },

    "Reliability-priority": {
        "reach": 0.10,
        "reliability": 0.55,
        "cost": 0.10,
        "delay": 0.10,
        "policy_risk": 0.15
    },

    "Delay-priority": {
        "reach": 0.10,
        "reliability": 0.10,
        "cost": 0.10,
        "delay": 0.55,
        "policy_risk": 0.15
    },

    "Policy-risk-priority": {
        "reach": 0.10,
        "reliability": 0.10,
        "cost": 0.10,
        "delay": 0.15,
        "policy_risk": 0.55
    }
}

# Verify that every configuration sums to one.
for config_name, weights in weight_configurations.items():

    weight_sum = sum(weights.values())

    if not np.isclose(weight_sum, 1.0):
        raise ValueError(
            f"Weights for {config_name} sum to {weight_sum}, "
            "not 1.0."
        )

# ------------------------------------------------------------
# 5. DISPLAY WEIGHT CONFIGURATIONS
# ------------------------------------------------------------

weight_table = pd.DataFrame(
    weight_configurations
).T

weight_table.index.name = "Weight configuration"

print("\nWEIGHT CONFIGURATIONS")
display(weight_table.round(3))

weight_table.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "objective_weight_configurations.csv"
    )
)

# ------------------------------------------------------------
# 6. PREPARE GRAPH HEURISTIC
# ------------------------------------------------------------

nodes = sorted(G_refined.nodes())

node_to_index = {
    node: index
    for index, node in enumerate(nodes)
}

n_nodes = len(nodes)

similarity_matrix = np.zeros(
    (n_nodes, n_nodes),
    dtype=float
)

for u, v, edge_data in G_refined.edges(data=True):

    similarity = float(
        edge_data.get(
            "similarity",
            edge_data.get("weight", 0.0)
        )
    )

    if not np.isfinite(similarity):
        similarity = 0.0

    similarity = max(similarity, 0.0)

    i = node_to_index[u]
    j = node_to_index[v]

    similarity_matrix[i, j] = similarity
    similarity_matrix[j, i] = similarity

# Normalize heuristic values across valid edges.
valid_edge_mask = similarity_matrix > 0

if not valid_edge_mask.any():
    raise ValueError(
        "No positive edge-similarity values are available."
    )

edge_similarities = similarity_matrix[valid_edge_mask]

sim_min = edge_similarities.min()
sim_max = edge_similarities.max()

if sim_max > sim_min:
    similarity_matrix[valid_edge_mask] = (
        (similarity_matrix[valid_edge_mask] - sim_min)
        / (sim_max - sim_min)
    )
else:
    similarity_matrix[valid_edge_mask] = 1.0

# Prevent zero desirability on valid edges.
similarity_matrix[valid_edge_mask] = np.maximum(
    similarity_matrix[valid_edge_mask],
    1e-8
)

# ------------------------------------------------------------
# 7. OBJECTIVE FUNCTION
# ------------------------------------------------------------

def calculate_weighted_objective(path, weights):
    """
    Calculate the mean node-level objective score.

    All objective components use normalized values.
    Higher reach and reliability are beneficial.
    Lower cost, delay, and policy risk are beneficial.
    """

    if path is None or len(path) == 0:
        return 0.0

    positions = np.asarray(path, dtype=int)

    path_data = normalized_df.iloc[positions]

    mean_reach = float(path_data["reach"].mean())

    mean_reliability = float(
        path_data["reliability"].mean()
    )

    mean_cost_benefit = float(
        path_data["cost_benefit"].mean()
    )

    mean_delay_benefit = float(
        path_data["delay_benefit"].mean()
    )

    mean_risk_benefit = float(
        path_data["risk_benefit"].mean()
    )

    score = (
        weights["reach"] * mean_reach
        + weights["reliability"] * mean_reliability
        + weights["cost"] * mean_cost_benefit
        + weights["delay"] * mean_delay_benefit
        + weights["policy_risk"] * mean_risk_benefit
    )

    return float(score)


# ------------------------------------------------------------
# 8. ANT PATH CONSTRUCTION
# ------------------------------------------------------------

def construct_sensitivity_path(
    start_node,
    pheromone,
    weights,
    rng
):
    """
    Construct a feasible simple path using current pheromone
    and the graph-similarity heuristic.
    """

    path = [int(start_node)]
    visited = {int(start_node)}

    while len(path) < min(PATH_LENGTH, n_nodes):

        current_node = path[-1]

        candidates = [
            int(neighbor)
            for neighbor in G_refined.neighbors(current_node)
            if int(neighbor) not in visited
        ]

        if not candidates:
            break

        current_index = node_to_index[current_node]

        desirabilities = []

        for candidate in candidates:

            candidate_index = node_to_index[candidate]

            tau = max(
                pheromone[current_index, candidate_index],
                1e-12
            )

            eta = max(
                similarity_matrix[
                    current_index,
                    candidate_index
                ],
                1e-12
            )

            desirability = (
                tau ** ALPHA
            ) * (
                eta ** BETA
            )

            desirabilities.append(desirability)

        desirabilities = np.asarray(
            desirabilities,
            dtype=float
        )

        total = desirabilities.sum()

        if (
            not np.isfinite(total)
            or total <= 0
        ):
            probabilities = np.full(
                len(candidates),
                1.0 / len(candidates)
            )
        else:
            probabilities = desirabilities / total

        next_node = int(
            rng.choice(
                candidates,
                p=probabilities
            )
        )

        path.append(next_node)
        visited.add(next_node)

    return path


# ------------------------------------------------------------
# 9. RUN DYNACO FOR ONE WEIGHT CONFIGURATION
# ------------------------------------------------------------

def run_weight_sensitivity(config_name, weights, seed):
    """
    Independently optimize a pathway under one weight
    configuration.
    """

    rng = np.random.default_rng(seed)

    pheromone = np.zeros(
        (n_nodes, n_nodes),
        dtype=float
    )

    # Initialize pheromone only on existing graph edges.
    pheromone[valid_edge_mask] = INITIAL_PHEROMONE

    global_best_path = None
    global_best_score = -np.inf

    iteration_history = []

    start_time = time.perf_counter()

    for iteration in range(1, N_ITERATIONS + 1):

        candidate_paths = []
        candidate_scores = []

        # -----------------------------------------------
        # Generate ant pathways
        # -----------------------------------------------

        for ant in range(N_ANTS):

            start_node = int(
                rng.choice(nodes)
            )

            path = construct_sensitivity_path(
                start_node=start_node,
                pheromone=pheromone,
                weights=weights,
                rng=rng
            )

            # Evaluate paths of at least two nodes.
            if len(path) < 2:
                continue

            score = calculate_weighted_objective(
                path,
                weights
            )

            candidate_paths.append(path)
            candidate_scores.append(score)

            if score > global_best_score:

                global_best_score = score
                global_best_path = path.copy()

        if not candidate_paths:
            raise RuntimeError(
                f"No valid candidate paths generated for "
                f"{config_name}, iteration {iteration}."
            )

        # -----------------------------------------------
        # Record iteration results
        # -----------------------------------------------

        best_iteration_index = int(
            np.argmax(candidate_scores)
        )

        iteration_best_path = candidate_paths[
            best_iteration_index
        ]

        iteration_best_score = candidate_scores[
            best_iteration_index
        ]

        iteration_history.append({
            "iteration": iteration,
            "best_score": float(iteration_best_score),
            "mean_score": float(
                np.mean(candidate_scores)
            ),
            "global_best_score": float(
                global_best_score
            )
        })

        # -----------------------------------------------
        # Dynamic pheromone evaporation
        # -----------------------------------------------

        evaporation_rate = (
            EVAPORATION_MAX
            - (
                np.clip(
                    global_best_score,
                    0.0,
                    1.0
                )
                * (
                    EVAPORATION_MAX
                    - EVAPORATION_MIN
                )
            )
        )

        evaporation_rate = float(
            np.clip(
                evaporation_rate,
                EVAPORATION_MIN,
                EVAPORATION_MAX
            )
        )

        pheromone[valid_edge_mask] *= (
            1.0 - evaporation_rate
        )

        # -----------------------------------------------
        # Deposit pheromone for candidate paths
        # -----------------------------------------------

        for path, score in zip(
            candidate_paths,
            candidate_scores
        ):

            deposit = max(
                float(score),
                1e-8
            )

            for u, v in zip(
                path[:-1],
                path[1:]
            ):

                i = node_to_index[u]
                j = node_to_index[v]

                pheromone[i, j] += deposit
                pheromone[j, i] += deposit

        # -----------------------------------------------
        # Elitist reinforcement
        # -----------------------------------------------

        elite_deposit = (
            ELITE_REINFORCEMENT
            * max(global_best_score, 1e-8)
        )

        for u, v in zip(
            global_best_path[:-1],
            global_best_path[1:]
        ):

            i = node_to_index[u]
            j = node_to_index[v]

            pheromone[i, j] += elite_deposit
            pheromone[j, i] += elite_deposit

        # Keep pheromone values finite and nonnegative.
        pheromone[~np.isfinite(pheromone)] = 0.0

        pheromone[valid_edge_mask] = np.clip(
            pheromone[valid_edge_mask],
            1e-12,
            1e6
        )

        # Non-edges must remain unavailable.
        pheromone[~valid_edge_mask] = 0.0

    elapsed_time = time.perf_counter() - start_time

    # --------------------------------------------------------
    # Recalculate final score independently
    # --------------------------------------------------------

    final_score = calculate_weighted_objective(
        global_best_path,
        weights
    )

    final_data = normalized_df.iloc[
        np.asarray(global_best_path, dtype=int)
    ]

    final_components = {
        "Reach": float(final_data["reach"].mean()),
        "Reliability": float(
            final_data["reliability"].mean()
        ),
        "Cost benefit": float(
            final_data["cost_benefit"].mean()
        ),
        "Delay benefit": float(
            final_data["delay_benefit"].mean()
        ),
        "Policy-risk benefit": float(
            final_data["risk_benefit"].mean()
        )
    }

    return {
        "configuration": config_name,
        "weights": weights.copy(),
        "objective_score": float(final_score),
        "path": global_best_path.copy(),
        "path_length": len(global_best_path),
        "runtime_seconds": float(elapsed_time),
        "components": final_components,
        "history": iteration_history
    }


# ------------------------------------------------------------
# 10. RUN ALL SIX CONFIGURATIONS
# ------------------------------------------------------------

sensitivity_results = []

for config_index, (
    config_name,
    weights
) in enumerate(weight_configurations.items()):

    print(f"\nConfiguration: {config_name}")

    result = run_weight_sensitivity(
        config_name=config_name,
        weights=weights,
        seed=SEED + config_index
    )

    sensitivity_results.append(result)

    print(
        f"Objective score : "
        f"{result['objective_score']:.6f}"
    )

    print(
        f"Path length     : "
        f"{result['path_length']}"
    )

    print(
        f"Runtime         : "
        f"{result['runtime_seconds']:.3f} seconds"
    )

    print(
        f"Selected path   : "
        f"{result['path']}"
    )

# ------------------------------------------------------------
# 11. OBJECTIVE-SCORE TABLE
# ------------------------------------------------------------

summary_rows = []

for result in sensitivity_results:

    summary_rows.append({
        "Weight configuration": result["configuration"],
        "Objective score": result["objective_score"],
        "Path length": result["path_length"],
        "Runtime (seconds)": result["runtime_seconds"],
        "Selected pathway": str(result["path"])
    })

sensitivity_summary_df = pd.DataFrame(
    summary_rows
)

sensitivity_summary_df = (
    sensitivity_summary_df.sort_values(
        "Objective score",
        ascending=False
    ).reset_index(drop=True)
)

print("\nOBJECTIVE-SCORE SENSITIVITY RESULTS")
display(
    sensitivity_summary_df.style.format({
        "Objective score": "{:.6f}",
        "Runtime (seconds)": "{:.3f}"
    })
)

sensitivity_summary_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "objective_score_sensitivity.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 12. COMPONENT-LEVEL COMPARISON TABLE
# ------------------------------------------------------------

component_rows = []

for result in sensitivity_results:

    component_rows.append({
        "Weight configuration": result["configuration"],
        "Objective score": result["objective_score"],
        "Mean reach": result["components"]["Reach"],
        "Mean reliability": result["components"]["Reliability"],
        "Mean cost benefit": result["components"]["Cost benefit"],
        "Mean delay benefit": result["components"]["Delay benefit"],
        "Mean policy-risk benefit": (
            result["components"]["Policy-risk benefit"]
        )
    })

component_summary_df = pd.DataFrame(
    component_rows
)

print("\nPATHWAY COMPONENT COMPARISON")
display(
    component_summary_df.round(5)
)

component_summary_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "pathway_component_sensitivity.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# 16. SAVE WEIGHTED COMPONENT CONTRIBUTIONS
# ------------------------------------------------------------

contribution_rows = []

for result in sensitivity_results:

    weights = result["weights"]
    components = result["components"]

    contribution_rows.append({
        "Weight configuration": result["configuration"],
        "Reach contribution": (
            weights["reach"] * components["Reach"]
        ),
        "Reliability contribution": (
            weights["reliability"]
            * components["Reliability"]
        ),
        "Cost contribution": (
            weights["cost"]
            * components["Cost benefit"]
        ),
        "Delay contribution": (
            weights["delay"]
            * components["Delay benefit"]
        ),
        "Policy-risk contribution": (
            weights["policy_risk"]
            * components["Policy-risk benefit"]
        ),
        "Recalculated objective score": (
            calculate_weighted_objective(
                result["path"],
                weights
            )
        )
    })

contribution_df = pd.DataFrame(
    contribution_rows
)

contribution_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "weighted_objective_contributions.csv"
    ),
    index=False
)

print("\nWEIGHTED OBJECTIVE CONTRIBUTIONS")
display(contribution_df.round(6))

# ------------------------------------------------------------
# 17. VALIDATE THE RESULTS
# ------------------------------------------------------------

print("\nVALIDATION")
print("-" * 75)

for result in sensitivity_results:

    path = result["path"]

    valid_edges = all(
        G_refined.has_edge(u, v)
        for u, v in zip(path[:-1], path[1:])
    )

    unique_nodes = (
        len(path) == len(set(path))
    )

    recalculated_score = calculate_weighted_objective(
        path,
        result["weights"]
    )

    score_consistent = np.isclose(
        recalculated_score,
        result["objective_score"],
        atol=1e-10
    )

    print(f"\n{result['configuration']}")

    print(f"Valid graph edges : {valid_edges}")
    print(f"Unique nodes      : {unique_nodes}")
    print(f"Score consistent  : {score_consistent}")
    print(f"Objective score   : {recalculated_score:.6f}")

# ------------------------------------------------------------
# 18. FINAL OUTPUT LOCATIONS
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("SENSITIVITY ANALYSIS OF OBJECTIVE-FUNCTION WEIGHTS IN DynACO-WG-SAGE")
print("=" * 75)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

OUTPUT_DIR = "Five_Fold_Cross_Validation_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

RANDOM_STATE = 42
N_SPLITS = 5

TARGET_COLUMN = "selected_path"

# ------------------------------------------------------------
# 2. VALIDATE DATASET
# ------------------------------------------------------------

if "df" not in globals():
    raise NameError(
        "Dataset 'df' is not available. "
        "Run the dataset-loading cell first."
    )

if TARGET_COLUMN not in df.columns:
    raise ValueError(
        f"Target column '{TARGET_COLUMN}' was not found. "
        f"Available columns: {df.columns.tolist()}"
    )

data = df.copy()

# Exclude identifiers and target-related leakage columns.
# Add other known target-derived columns to this list.

excluded_columns = [
    TARGET_COLUMN,
    "node_id",
    "selected_path_prediction",
    "predicted_selected_path"
]

excluded_columns = [
    col for col in excluded_columns
    if col in data.columns
]

X = data.drop(
    columns=excluded_columns
).copy()

y = data[TARGET_COLUMN].copy()

# Retain numeric predictors in this example.
X = X.select_dtypes(include=[np.number])

if X.shape[1] == 0:
    raise ValueError(
    )

# Remove constant columns.
constant_columns = [
    col for col in X.columns
    if X[col].nunique(dropna=True) <= 1
]

X = X.drop(columns=constant_columns)

if X.shape[1] == 0:
    raise ValueError(

    )

if y.isna().any():
    raise ValueError(
    )

if y.nunique() < 2:
    raise ValueError(
    )

if not all(
    pd.api.types.is_numeric_dtype(X[col])
    for col in X.columns
):
    raise ValueError(
    )

if not np.isfinite(
    X.to_numpy(dtype=float)
).all():
    # Missing values are allowed and imputed within each fold,
    # but infinite values must be resolved first.
    X = X.replace(
        [np.inf, -np.inf],
        np.nan
    )

class_counts = y.value_counts()

if class_counts.min() < N_SPLITS:
    raise ValueError(
    )

# ------------------------------------------------------------
# 3. DEFINE CROSS-VALIDATION FOLDS
# ------------------------------------------------------------

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

fold_indices = list(
    cv.split(X, y)
)

# ------------------------------------------------------------
# 4. DEFINE BASELINE CLASSIFIERS
# ------------------------------------------------------------

rf_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "classifier",
        RandomForestClassifier(
            n_estimators=200,
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1
        )
    )
])

svm_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        SVC(
            kernel="rbf",
            C=1.0,
            gamma="scale",
            class_weight="balanced"
        )
    )
])

baseline_models = {
    "Random Forest": rf_model,
    "SVM": svm_model
}

# ------------------------------------------------------------
# 5. DEFINE GRAPH-MODEL ADAPTERS
# ------------------------------------------------------------

def predict_acgnn_fold(
    X_train,
    y_train,
    X_test,
    train_indices,
    test_indices
):

    raise NotImplementedError(
    )


def predict_dynaco_wg_sage_fold(
    X_train,
    y_train,
    X_test,
    train_indices,
    test_indices
):

    raise NotImplementedError()


graph_model_adapters = {
    "ACGNN": predict_acgnn_fold,
    "DynACO-WG-SAGE (Proposed)": predict_dynaco_wg_sage_fold
}

# ------------------------------------------------------------
# 6. RUN FIVE-FOLD CROSS-VALIDATION
# ------------------------------------------------------------

results = {
    model_name: []
    for model_name in [
        "ACGNN",
        "Random Forest",
        "SVM",
        "DynACO-WG-SAGE (Proposed)"
    ]
}

for fold_number, (
    train_idx,
    test_idx
) in enumerate(fold_indices, start=1):

    print(f"\n{'-' * 55}")
    print(f"FOLD {fold_number}")
    print(f"{'-' * 55}")

    X_train = X.iloc[train_idx].copy()
    X_test = X.iloc[test_idx].copy()

    y_train = y.iloc[train_idx].copy()
    y_test = y.iloc[test_idx].copy()

    # --------------------------------------------------------
    # A. RANDOM FOREST AND SVM
    # --------------------------------------------------------

    for model_name, model in baseline_models.items():

        fitted_model = clone(model)

        fitted_model.fit(
            X_train,
            y_train
        )

        predictions = fitted_model.predict(
            X_test
        )

        accuracy = accuracy_score(
            y_test,
            predictions
        ) * 100.0

        results[model_name].append(
            float(accuracy)
        )

        print(
            f"{model_name:30s}: "
            f"{accuracy:.2f}%"
        )

    # --------------------------------------------------------
    # B. ACGNN AND DYNACO-WG-SAGE
    # --------------------------------------------------------
    # These models cannot be evaluated until their actual
    # fold-specific training adapters are connected.

    for model_name, adapter in graph_model_adapters.items():

        try:
            predictions = adapter(
                X_train=X_train,
                y_train=y_train,
                X_test=X_test,
                train_indices=train_idx,
                test_indices=test_idx
            )

            predictions = np.asarray(
                predictions
            ).reshape(-1)

            if len(predictions) != len(y_test):
                raise ValueError(
                    f"{model_name} returned "
                    f"{len(predictions)} predictions for "
                    f"{len(y_test)} test records."
                )

            accuracy = accuracy_score(
                y_test,
                predictions
            ) * 100.0

            results[model_name].append(
                float(accuracy)
            )

            print(
                f"{model_name:30s}: "
                f"{accuracy:.2f}%"
            )

        except NotImplementedError as exc:
            print(
                f"{model_name}: NOT RUN — {exc}"
            )

# ------------------------------------------------------------
# 7. BUILD THE FIVE-FOLD TABLE
# ------------------------------------------------------------

rows = []

for model_name, fold_scores in results.items():

    if len(fold_scores) != N_SPLITS:
        rows.append({
            "Model": model_name,
            "Fold 1": np.nan,
            "Fold 2": np.nan,
            "Fold 3": np.nan,
            "Fold 4": np.nan,
            "Fold 5": np.nan,
            "Mean": np.nan,
            "SD": np.nan
        })
        continue

    rows.append({
        "Model": model_name,
        "Fold 1": fold_scores[0],
        "Fold 2": fold_scores[1],
        "Fold 3": fold_scores[2],
        "Fold 4": fold_scores[3],
        "Fold 5": fold_scores[4],
        "Mean": np.mean(fold_scores),
        "SD": np.std(fold_scores, ddof=1)
    })

cv_results_df = pd.DataFrame(
    rows
)

print("\n" + "=" * 90)
print("FIVE-FOLD CROSS-VALIDATION RESULTS")
print("Metric: Accuracy (%)")
print("=" * 90)

display(
    cv_results_df.style.format(
        {
            "Fold 1": "{:.2f}",
            "Fold 2": "{:.2f}",
            "Fold 3": "{:.2f}",
            "Fold 4": "{:.2f}",
            "Fold 5": "{:.2f}",
            "Mean": "{:.2f}",
            "SD": "{:.2f}"
        },
        na_rep="Not run"
    )
)

# ------------------------------------------------------------
# 8. SAVE CSV RESULTS
# ------------------------------------------------------------

csv_path = os.path.join(
    OUTPUT_DIR,
    "five_fold_cross_validation_results.csv"
)

cv_results_df.to_csv(
    csv_path,
    index=False
)

print("\nSaved results:")
print(os.path.abspath(csv_path))

# ------------------------------------------------------------
# 9. GENERATE HEATMAP
# ------------------------------------------------------------

fold_columns = [
    "Fold 1",
    "Fold 2",
    "Fold 3",
    "Fold 4",
    "Fold 5"
]

plot_df = cv_results_df.set_index(
    "Model"
)[fold_columns]

# Plot only models that completed all five folds.
plot_df = plot_df.dropna(
    how="any"
)

if not plot_df.empty:

    fig, ax = plt.subplots(
        figsize=(11, max(4, 1.1 * len(plot_df)))
    )

    matrix = plot_df.to_numpy()

    im = ax.imshow(
        matrix,
        aspect="auto",
        cmap="viridis",
        vmin=0,
        vmax=100
    )

    ax.set_xticks(
        np.arange(len(fold_columns))
    )

    ax.set_xticklabels(
        fold_columns
    )

    ax.set_yticks(
        np.arange(len(plot_df))
    )

    ax.set_yticklabels(
        plot_df.index
    )

    ax.set_title(
        "Five-Fold Cross-Validation Accuracy",
        fontsize=14,
        fontweight="bold",
        pad=12
    )

    ax.set_xlabel(
        "Cross-validation fold"
    )

    ax.set_ylabel(
        "Model"
    )

    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):

            ax.text(
                j,
                i,
                f"{matrix[i, j]:.2f}",
                ha="center",
                va="center",
                fontsize=10,
                color="white"
            )

    cbar = fig.colorbar(
        im,
        ax=ax
    )

    cbar.set_label(
        "Accuracy (%)"
    )

    plt.tight_layout()

    heatmap_path = os.path.join(
        OUTPUT_DIR,
        "five_fold_accuracy_heatmap.png"
    )

    fig.savefig(
        heatmap_path,
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()

    print(
        "Saved heatmap:",
        os.path.abspath(heatmap_path)
    )

else:
    print(
        "\nHeatmap not generated because no model has "
        "completed all five folds."
    )
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
from IPython.display import display

# ---------------------------------------------------------------
# 1. CONFIGURATION
# ---------------------------------------------------------------

OUTPUT_DIR = "DynACO_WG_SAGE_Statistical_Analysis"
os.makedirs(OUTPUT_DIR, exist_ok=True)

N_FOLDS = 5
DF = N_FOLDS - 1
ALPHA = 0.05

PROPOSED_MODEL = "DynACO-WG-SAGE (Proposed)"

BASELINE_MODELS = [
    "ACGNN",
    "Random Forest",
    "SVM"
]

FOLD_COLUMNS = [
    "Fold 1",
    "Fold 2",
    "Fold 3",
    "Fold 4",
    "Fold 5"
]


# ---------------------------------------------------------------
# 2. RETRIEVE FIVE-FOLD ACCURACY VALUES
# ---------------------------------------------------------------

fold_scores = {}

if "cv_results_df" in globals():

    cv_table = cv_results_df.copy()

    required_columns = ["Model"] + FOLD_COLUMNS

    missing = [
        col for col in required_columns
        if col not in cv_table.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns in cv_results_df: {missing}"
        )

    for model_name in [
        "ACGNN",
        "Random Forest",
        "SVM",
        "DynACO-WG-SAGE (Proposed)"
    ]:

        selected = cv_table.loc[
            cv_table["Model"] == model_name
        ]

        if selected.empty:
            continue

        fold_scores[model_name] = (
            selected.iloc[0][FOLD_COLUMNS]
            .to_numpy(dtype=float)
        )

elif "results" in globals():

    for model_name, scores in results.items():

        fold_scores[model_name] = np.asarray(
            scores,
            dtype=float
        )

else:

    raise NameError(
        "Cross-validation results were not found. "
        "Run the five-fold cross-validation cell first."
    )

# ---------------------------------------------------------------
# 3. VALIDATE THE FOLD SCORES
# ---------------------------------------------------------------

all_models = BASELINE_MODELS + [PROPOSED_MODEL]

for model_name in all_models:

    if model_name not in fold_scores:
        raise ValueError(
            f"Missing fold scores for '{model_name}'. "
            "Ensure all four models completed all five folds."
        )

    scores = np.asarray(
        fold_scores[model_name],
        dtype=float
    )

    if len(scores) != N_FOLDS:
        raise ValueError(
            f"{model_name} has {len(scores)} fold scores; "
            f"exactly {N_FOLDS} are required."
        )

    if not np.isfinite(scores).all():
        raise ValueError(
            f"{model_name} contains missing or non-finite "
            "fold scores."
        )

    if np.any((scores < 0) | (scores > 100)):
        raise ValueError(
            f"{model_name} scores must be percentages "
            "between 0 and 100."
        )

    fold_scores[model_name] = scores

print("\nFive-fold accuracy values (%)")

for model_name in all_models:
    print(
        f"{model_name:32s}: "
        f"{np.round(fold_scores[model_name], 4)}"
    )

# ---------------------------------------------------------------
# 4. COMPUTE MEAN, SD, T-TEST, AND CONFIDENCE INTERVAL
# ---------------------------------------------------------------

# Two-sided 95% confidence interval critical value.
t_critical = stats.t.ppf(
    1 - ALPHA / 2,
    df=DF
)

statistical_rows = []
paired_difference_rows = []

for baseline in BASELINE_MODELS:

    baseline_scores = fold_scores[baseline]

    proposed_scores = fold_scores[PROPOSED_MODEL]

    # -----------------------------------------------------------
    # Paired differences: proposed minus baseline
    # -----------------------------------------------------------

    differences = (
        proposed_scores - baseline_scores
    )

    mean_difference = float(
        np.mean(differences)
    )

    sd_difference = float(
        np.std(differences, ddof=1)
    )

    standard_error = (
        sd_difference / np.sqrt(N_FOLDS)
    )

    # -----------------------------------------------------------
    # Paired t-test
    # -----------------------------------------------------------

    if np.isclose(sd_difference, 0.0):

        if np.isclose(mean_difference, 0.0):

            t_value = 0.0
            p_value = 1.0

        else:

            # All fold differences are the same nonzero value.
            # The conventional t statistic is mathematically
            # infinite in magnitude.
            t_value = (
                np.inf
                if mean_difference > 0
                else -np.inf
            )

            p_value = 0.0

    else:

        t_value = (
            mean_difference / standard_error
        )

        p_value = float(
            stats.ttest_rel(
                proposed_scores,
                baseline_scores,
                alternative="two-sided"
            ).pvalue
        )

    # -----------------------------------------------------------
    # 95% confidence interval for mean paired difference
    # -----------------------------------------------------------

    if np.isclose(sd_difference, 0.0):

        ci_lower = mean_difference
        ci_upper = mean_difference

    else:

        margin = (
            t_critical * standard_error
        )

        ci_lower = (
            mean_difference - margin
        )

        ci_upper = (
            mean_difference + margin
        )

    # -----------------------------------------------------------
    # Descriptive statistics
    # -----------------------------------------------------------

    baseline_mean = float(
        np.mean(baseline_scores)
    )

    baseline_sd = float(
        np.std(
            baseline_scores,
            ddof=1
        )
    )

    proposed_mean = float(
        np.mean(proposed_scores)
    )

    proposed_sd = float(
        np.std(
            proposed_scores,
            ddof=1
        )
    )

    # -----------------------------------------------------------
    # Statistical decision
    # -----------------------------------------------------------

    significant = p_value < ALPHA

    if significant and mean_difference > 0:

        significance_label = (
            "Significant; proposed higher"
        )

    elif significant and mean_difference < 0:

        significance_label = (
            "Significant; proposed lower"
        )

    else:

        significance_label = (
            "Not significant"
        )

    comparison_name = (
        f"{baseline} Vs DynACO-WG-SAGE"
    )

    statistical_rows.append({

        "Baseline Methods": comparison_name,

        "Baseline Mean ± SD (%)": (
            f"{baseline_mean:.2f} ± {baseline_sd:.2f}"
        ),

        "DynACO-WG-SAGE Mean ± SD (%)": (
            f"{proposed_mean:.2f} ± {proposed_sd:.2f}"
        ),

        "Mean Diff": mean_difference,

        "t-value (df=4)": t_value,

        "p-value (< 0.05)": p_value,

        "95% CI of Diff": (
            f"[{ci_lower:.3f}, {ci_upper:.3f}]"
        ),

        "Significance (α=0.05)": significance_label

    })

    # Store individual fold differences for transparency.
    paired_difference_rows.append({

        "Baseline Methods": comparison_name,

        "Fold 1 Difference": differences[0],

        "Fold 2 Difference": differences[1],

        "Fold 3 Difference": differences[2],

        "Fold 4 Difference": differences[3],

        "Fold 5 Difference": differences[4],

        "Mean Difference": mean_difference,

        "SD of Difference": sd_difference,

        "Standard Error": standard_error,

        "CI Lower": ci_lower,

        "CI Upper": ci_upper,

        "t-value": t_value,

        "p-value": p_value

    })

# ---------------------------------------------------------------
# 5. MULTIPLE-COMPARISON CORRECTION
# ---------------------------------------------------------------

raw_p_values = np.array([
    row["p-value (< 0.05)"]
    for row in statistical_rows
])

number_of_tests = len(raw_p_values)

sort_indices = np.argsort(raw_p_values)

adjusted_sorted = np.empty(
    number_of_tests,
    dtype=float
)

running_max = 0.0

for rank, idx in enumerate(sort_indices):

    multiplier = number_of_tests - rank

    adjusted = min(
        1.0,
        multiplier * raw_p_values[idx]
    )

    running_max = max(
        running_max,
        adjusted
    )

    adjusted_sorted[rank] = running_max

# Map adjusted p-values back to original comparison order.
holm_adjusted = np.empty(
    number_of_tests,
    dtype=float
)

for rank, idx in enumerate(sort_indices):
    holm_adjusted[idx] = adjusted_sorted[rank]

for idx, row in enumerate(statistical_rows):

    row["Holm-adjusted p-value"] = (
        holm_adjusted[idx]
    )

    row["Holm significance (α=0.05)"] = (
        "Significant"
        if holm_adjusted[idx] < ALPHA
        else "Not significant"
    )

# ---------------------------------------------------------------
# 6. DISPLAY THE REQUESTED STATISTICAL TABLE
# ---------------------------------------------------------------

statistical_df = pd.DataFrame(
    statistical_rows
)

print("\n" + "=" * 125)
print("STATISTICAL SIGNIFICANCE RESULTS")
print("=" * 125)

display(
    statistical_df.style.format({
        "Mean Diff": "{:.4f}",
        "t-value (df=4)": "{:.4f}",
        "p-value (< 0.05)": "{:.6f}",
        "Holm-adjusted p-value": "{:.6f}"
    })
)

# Save complete table.
statistical_csv = os.path.join(
    OUTPUT_DIR,
    "statistical_significance_results.csv"
)

statistical_df.to_csv(
    statistical_csv,
    index=False
)

# ---------------------------------------------------------------
# 7. DISPLAY FOLD-LEVEL PAIRED DIFFERENCES
# ---------------------------------------------------------------

paired_df = pd.DataFrame(
    paired_difference_rows
)


display(
    paired_df.style.format({
        "Fold 1 Difference": "{:.4f}",
        "Fold 2 Difference": "{:.4f}",
        "Fold 3 Difference": "{:.4f}",
        "Fold 4 Difference": "{:.4f}",
        "Fold 5 Difference": "{:.4f}",
        "Mean Difference": "{:.4f}",
        "SD of Difference": "{:.4f}",
        "Standard Error": "{:.4f}",
        "CI Lower": "{:.4f}",
        "CI Upper": "{:.4f}",
        "t-value": "{:.4f}",
        "p-value": "{:.6f}"
    })
)

paired_csv = os.path.join(
    OUTPUT_DIR,
    "paired_fold_differences.csv"
)

paired_df.to_csv(
    paired_csv,
    index=False
)

# ---------------------------------------------------------------
# 9. INTERPRETATION SUMMARY
# ---------------------------------------------------------------

print("\n" + "=" * 85)
print("INTERPRETATION")
print("=" * 85)

for row in statistical_rows:

    print(
        f"\nComparison: {row['Baseline Methods']}"
    )

    print(
        f"Mean difference: "
        f"{row['Mean Diff']:.4f} percentage points"
    )

    print(
        f"t-value: {row['t-value (df=4)']:.4f}"
    )

    print(
        f"Raw p-value: "
        f"{row['p-value (< 0.05)']:.6f}"
    )

    print(
        f"95% CI: {row['95% CI of Diff']}"
    )

    print(
        f"Raw significance: "
        f"{row['Significance (α=0.05)']}"
    )

    print(
        f"Holm-adjusted p-value: "
        f"{row['Holm-adjusted p-value']:.6f}"
    )

    print(
        f"Holm-adjusted decision: "
        f"{row['Holm significance (α=0.05)']}"
    )
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# ------------------------------------------------
# 1. OUTPUT DIRECTORY
# ------------------------------------------------
OUTPUT_DIR = "DynACO_WG_SAGE_Pathway_Comparison"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------
# 2. CONFIGURATION
# ------------------------------------------------
# Threshold levels requested for the comparison.
THRESHOLD_LEVELS = np.round(np.linspace(0.0, 1.0, 6), 2)

PLANNING_CRITERIA = {
    "Cost": {
        "direction": "lower_is_better",
        "description": "Pathway cost"
    },
    "Audience reach": {
        "direction": "higher_is_better",
        "description": "Expected audience reach"
    },
    "Reliability": {
        "direction": "higher_is_better",
        "description": "Pathway reliability"
    },
    "Time delay": {
        "direction": "lower_is_better",
        "description": "Time required for dissemination"
    },
    "Policy risk": {
        "direction": "lower_is_better",
        "description": "Policy-related pathway risk"
    },
    "optimal_path_score": {
        "direction": "higher_is_better",
        "description": "Overall pathway optimization score"
    },
    "selected_path": {
        "direction": "binary",
        "description": "Whether a pathway is selected"
    }
}

# ------------------------------------------------
# 3. INSERT ACTUAL MODEL OUTPUTS
# ------------------------------------------------

baseline_metrics = {
    "Cost": 0.62,
    "Audience reach": 0.68,
    "Reliability": 0.72,
    "Time delay": 0.55,
    "Policy risk": 0.60,
    "optimal_path_score": 0.65,
    "selected_path": 1
}

proposed_metrics = {
    "Cost": 0.80,
    "Audience reach": 0.88,
    "Reliability": 0.91,
    "Time delay": 0.82,
    "Policy risk": 0.84,
    "optimal_path_score": 0.89,
    "selected_path": 1
}

# ------------------------------------------------
# 4. VALIDATE INPUTS
# ------------------------------------------------
def validate_metrics(metrics, method_name):
    required = list(PLANNING_CRITERIA.keys())

    missing = [c for c in required if c not in metrics]
    if missing:
        raise ValueError(
            f"{method_name} is missing the following criteria: {missing}"
        )

    for criterion in required:
        value = metrics[criterion]

        if not np.isfinite(value):
            raise ValueError(
                f"{method_name}: {criterion} must be finite."
            )

        if criterion == "selected_path":
            if value not in (0, 1):
                raise ValueError(
                    f"{method_name}: selected_path must be 0 or 1."
                )
        elif not 0 <= value <= 1:
            raise ValueError(
                f"{method_name}: {criterion} must be normalized to [0, 1]."
            )

validate_metrics(baseline_metrics, "Baseline")
validate_metrics(proposed_metrics, "DynACO-WG-SAGE")

# ------------------------------------------------
# 5. GENERATE THRESHOLD-BASED COMPARISON
# ------------------------------------------------

records = []

for criterion, metadata in PLANNING_CRITERIA.items():

    baseline_value = float(baseline_metrics[criterion])
    proposed_value = float(proposed_metrics[criterion])

    for threshold in THRESHOLD_LEVELS:

        if metadata["direction"] == "binary":

            baseline_indicator = int(baseline_value)
            proposed_indicator = int(proposed_value)

        else:

            baseline_indicator = int(
                baseline_value >= threshold
            )

            proposed_indicator = int(
                proposed_value >= threshold
            )

        # Absolute difference between binary indicators.
        # +1 = proposed succeeds where baseline fails.
        #  0 = both methods have the same indicator.
        # -1 = baseline succeeds where proposed fails.
        improvement = proposed_indicator - baseline_indicator

        records.append({
            "Planning criterion": criterion,
            "Threshold level (0–1)": float(threshold),
            "Baseline pathway (0/1)": baseline_indicator,
            "DynACO-WG-SAGE proposed pathway (0/1)": proposed_indicator,
            "Improvement": improvement,
            "Baseline desirability": baseline_value,
            "Proposed desirability": proposed_value
        })

comparison_df = pd.DataFrame(records)

# ------------------------------------------------
# 6. FORMAT RESULTS
# ------------------------------------------------
comparison_df["Threshold level (0–1)"] = (
    comparison_df["Threshold level (0–1)"].round(2)
)

comparison_df["Baseline desirability"] = (
    comparison_df["Baseline desirability"].round(4)
)

comparison_df["Proposed desirability"] = (
    comparison_df["Proposed desirability"].round(4)
)

print("\n" + "=" * 90)
print("COMPARISON OF CULTURAL DISSEMINATION PATHWAY PLANNING")
print("Baseline vs. Proposed DynACO-WG-SAGE")
print("=" * 90)

display(comparison_df)

# ------------------------------------------------
# 7. EXPORT DETAILED COMPARISON TABLE
# ------------------------------------------------
comparison_path = os.path.join(
    OUTPUT_DIR,
    "DynACO_WG_SAGE_Pathway_Comparison.csv"
)

comparison_df.to_csv(comparison_path, index=False)

print(f"\nDetailed comparison saved to:\n{comparison_path}")

# ------------------------------------------------
# 8. CREATE REQUESTED COMPACT TABLE
# ------------------------------------------------
compact_df = comparison_df[
    [
        "Planning criterion",
        "Threshold level (0–1)",
        "Baseline pathway (0/1)",
        "DynACO-WG-SAGE proposed pathway (0/1)",
        "Improvement"
    ]
].copy()

compact_path = os.path.join(
    OUTPUT_DIR,
    "DynACO_WG_SAGE_Compact_Comparison.csv"
)

compact_df.to_csv(compact_path, index=False)

print(f"Compact table saved to:\n{compact_path}")

# ------------------------------------------------
# 9. SUMMARIZE INDICATOR-LEVEL PERFORMANCE
# ------------------------------------------------
summary_df = (
    comparison_df
    .groupby("Planning criterion", sort=False)
    .agg(
        Baseline_Success_Rate=("Baseline pathway (0/1)", "mean"),
        Proposed_Success_Rate=(
            "DynACO-WG-SAGE proposed pathway (0/1)", "mean"
        ),
        Mean_Indicator_Improvement=("Improvement", "mean")
    )
    .reset_index()
)

summary_df["Baseline_Success_Rate"] = (
    summary_df["Baseline_Success_Rate"] * 100
)

summary_df["Proposed_Success_Rate"] = (
    summary_df["Proposed_Success_Rate"] * 100
)

summary_df["Mean_Indicator_Improvement"] = (
    summary_df["Mean_Indicator_Improvement"] * 100
)

summary_df = summary_df.round(2)

print("\nThreshold-Based Summary (%):")
display(summary_df)

summary_path = os.path.join(
    OUTPUT_DIR,
    "DynACO_WG_SAGE_Threshold_Summary.csv"
)

summary_df.to_csv(summary_path, index=False)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
from IPython.display import display

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

OUTPUT_DIR = "Complex_Cultural_Network_Evaluation"
os.makedirs(OUTPUT_DIR, exist_ok=True)

required_columns = [
    "cost",
    "reach",
    "reliability",
    "time_delay",
    "policy_risk",
    "optimal_path_score",
    "selected_path"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(
        f"Required columns are missing from df: {missing_columns}"
    )

# Use the existing notebook dataset
evaluation_df = df.copy()

# Remove records with missing evaluation values
evaluation_df = evaluation_df.dropna(subset=required_columns)

if evaluation_df.empty:
    raise ValueError("No valid records remain for network evaluation.")

# ------------------------------------------------------------
# 2. BUILD THE CULTURAL DISSEMINATION NETWORK
# ------------------------------------------------------------

G_eval = nx.Graph()

for idx, row in evaluation_df.iterrows():
    node_id = str(idx)

    G_eval.add_node(
        node_id,
        event_type=str(row["event_type"]),
        platform=str(row["platform"]),
        audience=str(row["audience"]),
        cost=float(row["cost"]),
        reach=float(row["reach"]),
        reliability=float(row["reliability"]),
        time_delay=float(row["time_delay"]),
        policy_risk=float(row["policy_risk"]),
        optimal_path_score=float(row["optimal_path_score"]),
        selected_path=int(row["selected_path"])
    )

# Connect records sharing a dissemination platform or audience.
records = evaluation_df.reset_index(drop=True)

for i in range(len(records)):
    for j in range(i + 1, len(records)):
        same_platform = (
            records.loc[i, "platform"] == records.loc[j, "platform"]
        )
        same_audience = (
            records.loc[i, "audience"] == records.loc[j, "audience"]
        )

        if same_platform or same_audience:
            G_eval.add_edge(str(evaluation_df.index[i]),
                            str(evaluation_df.index[j]))

# ------------------------------------------------------------
# 3. NETWORK METRIC CALCULATION
# ------------------------------------------------------------

# Candidate pathways evaluated:
# Number of records available as candidate dissemination pathways.
candidate_pathways = len(evaluation_df)

component_path_lengths = []

for component_nodes in nx.connected_components(G_eval):
    subgraph = G_eval.subgraph(component_nodes)

    if subgraph.number_of_nodes() > 1:
        component_path_lengths.append(
            (
                nx.average_shortest_path_length(subgraph),
                subgraph.number_of_nodes()
            )
        )

if component_path_lengths:
    total_pairs_weight = sum(
        size * (size - 1)
        for _, size in component_path_lengths
    )

    average_path_length = (
        sum(
            length * size * (size - 1)
            for length, size in component_path_lengths
        ) / total_pairs_weight
    )
else:
    average_path_length = np.nan

# Pathway reliability:
# Mean reliability among selected pathways.
selected_df = evaluation_df[
    evaluation_df["selected_path"].astype(int) == 1
]

if not selected_df.empty:
    pathway_reliability = selected_df["reliability"].mean()
else:
    pathway_reliability = np.nan

# Network coverage:
# Percentage of unique audiences represented by selected pathways.
all_audiences = set(evaluation_df["audience"].astype(str).unique())

if all_audiences:
    covered_audiences = set(selected_df["audience"].astype(str).unique())

    network_coverage = (
        len(covered_audiences) / len(all_audiences)
    ) * 100
else:
    network_coverage = np.nan

# Pathway selection objective score:
# Mean optimal_path_score among selected pathways.
if not selected_df.empty:
    objective_score = selected_df["optimal_path_score"].mean()
else:
    objective_score = np.nan

# ------------------------------------------------------------
# 4. BASELINE AND PROPOSED METHOD COMPARISON
# ------------------------------------------------------------

evaluation_results = pd.DataFrame({
    "Network metric": [
        "Candidate pathways evaluated",
        "Average path length",
        "Pathway reliability",
        "Network coverage (%)",
        "Pathway selection objective score"
    ],
    "Dataset-based result": [
        candidate_pathways,
        average_path_length,
        pathway_reliability,
        network_coverage,
        objective_score
    ],
    "Preferred direction": [
        "Report search space; not inherently higher",
        "Lower",
        "Higher",
        "Higher",
        "Higher"
    ]
})

print("\n" + "=" * 75)
print("EVALUATION OF COMPLEX CULTURAL DISSEMINATION NETWORKS")
print("=" * 75)

display(evaluation_results.round(4))

# Save the dataset-derived evaluation table
evaluation_results.to_csv(
    os.path.join(OUTPUT_DIR, "complex_network_evaluation.csv"),
    index=False
)

# ------------------------------------------------------------
# 5. SELECTED VS NON-SELECTED PATHWAY ANALYSIS
# ------------------------------------------------------------

grouped_results = (
    evaluation_df
    .groupby("selected_path")
    .agg(
        Number_of_records=("selected_path", "size"),
        Mean_cost=("cost", "mean"),
        Mean_reach=("reach", "mean"),
        Mean_reliability=("reliability", "mean"),
        Mean_time_delay=("time_delay", "mean"),
        Mean_policy_risk=("policy_risk", "mean"),
        Mean_objective_score=("optimal_path_score", "mean")
    )
    .rename(index={0: "Not selected", 1: "Selected"})
)

print("\nSELECTED VS NON-SELECTED PATHWAY ANALYSIS")
display(grouped_results.round(4))

grouped_results.to_csv(
    os.path.join(OUTPUT_DIR, "selected_vs_nonselected_pathways.csv")
)
```
